# Genotox subprocess launcher v4

이 노트북은 **실제 실행되는 스크립트 버전과 해시를 먼저 확인**하고,
**실험을 endpoint/strategy/model 단위의 개별 subprocess**로 실행합니다.
그래서 한 실험이 실패해도 전체가 같이 죽지 않고, 어떤 조합이 실패했는지 바로 알 수 있습니다.

```python

from pathlib import Path
import json, shutil, subprocess, sys, hashlib, itertools, os
import pandas as pd
from IPython.display import display

BASE_DIR = Path(r"C:\Users\Administrator\Documents\GitHub\Genotox_model\model")
DATA_DIR = BASE_DIR / "data"
RUNS_DIR = BASE_DIR / "runs"
RUNTIME_DIR = BASE_DIR / "_notebook_runtime"
RUNTIME_DIR.mkdir(parents=True, exist_ok=True)

# 중요: 현재 노트북과 같은 폴더의 스크립트를 1순위로 사용
SCRIPT_CANDIDATES = [
    Path.cwd() / "genotox_subprocess_pipeline_patched_v4.py",
    BASE_DIR / "genotox_subprocess_pipeline_patched_v4.py",
]
SCRIPT_SRC = next((p for p in SCRIPT_CANDIDATES if p.exists()), None)
if SCRIPT_SRC is None:
    raise FileNotFoundError("genotox_subprocess_pipeline_patched_v4.py 파일을 찾지 못했습니다. 노트북과 같은 폴더 또는 model 폴더에 두세요.")

SCRIPT_DST = RUNTIME_DIR / "genotox_subprocess_pipeline.py"
shutil.copy2(SCRIPT_SRC, SCRIPT_DST)

sha = hashlib.sha256(SCRIPT_DST.read_bytes()).hexdigest()[:16]
head = "\n".join(SCRIPT_DST.read_text(encoding="utf-8", errors="ignore").splitlines()[:5])
print("SCRIPT_SRC:", SCRIPT_SRC)
print("SCRIPT_DST:", SCRIPT_DST)
print("SHA256[:16]:", sha)
print("--- script head ---")
print(head)

BASE_CONFIG = {
    "base_dir": str(BASE_DIR),
    "data_dir": str(DATA_DIR),
    "runs_dir": str(RUNS_DIR),
    "endpoints": ["ames", "invitro", "invivo"],
    "strategies": ["none", "alert_bootstrap"],
    "models": ["logreg", "xgb"],
    "feature_blocks": ["physchem", "rules", "fg_present", "fg_count", "fingerprint", "qm"],
    "fp_bits": 256,
    "fp_top_k": 128,
    "qm_file": None,
    "qm_merge_key": "No",
    "cv_splits": 3,
    "random_state": 42,
    "tuning_metric": "mcc",
    "search_n_iter": 12,
    "search_n_iter_logreg": 12,
    "search_n_iter_xgb": 8,
    "shap_background_n": 200,
    "shap_explain_n": 300,
    "force_rebuild_broadfp": True,
}
print(json.dumps(BASE_CONFIG, ensure_ascii=False, indent=2))
```

```python

def run_logged(cmd, log_path: Path):
    log_path.parent.mkdir(parents=True, exist_ok=True)
    with open(log_path, "w", encoding="utf-8", errors="ignore") as log:
        log.write("CMD: " + " ".join(cmd) + "\n\n")
        log.flush()
        proc = subprocess.run(cmd, stdout=log, stderr=log, text=True)
    tail = "\n".join(log_path.read_text(encoding="utf-8", errors="ignore").splitlines()[-80:])
    print("returncode =", proc.returncode)
    print("log file:", log_path)
    print("---- log tail ----")
    print(tail)
    return proc.returncode

# 1) Build once
build_cfg = dict(BASE_CONFIG)
build_cfg["force_rebuild_broadfp"] = True
BUILD_CFG_PATH = RUNTIME_DIR / "pipeline_build_config.json"
BUILD_CFG_PATH.write_text(json.dumps(build_cfg, ensure_ascii=False, indent=2), encoding="utf-8")
BUILD_LOG = RUNTIME_DIR / "build_subprocess_v4.log"
cmd = [sys.executable, str(SCRIPT_DST), "--config-json", str(BUILD_CFG_PATH), "--mode", "build"]
rc = run_logged(cmd, BUILD_LOG)
if rc != 0:
    raise RuntimeError("Build step failed. See build log above.")
```

```python

# 2) Run each experiment separately
combos = list(itertools.product(BASE_CONFIG["endpoints"], BASE_CONFIG["strategies"], BASE_CONFIG["models"]))
results = []
for endpoint, strategy, model in combos:
    cfg = dict(BASE_CONFIG)
    cfg["endpoints"] = [endpoint]
    cfg["strategies"] = [strategy]
    cfg["models"] = [model]
    cfg["force_rebuild_broadfp"] = False  # build already done
    cfg_path = RUNTIME_DIR / f"cfg__{endpoint}__{strategy}__{model}.json"
    cfg_path.write_text(json.dumps(cfg, ensure_ascii=False, indent=2), encoding="utf-8")
    log_path = RUNTIME_DIR / f"run__{endpoint}__{strategy}__{model}.log"
    print(f"\n=== RUN {endpoint} | {strategy} | {model} ===")
    rc = run_logged([sys.executable, str(SCRIPT_DST), "--config-json", str(cfg_path), "--mode", "run"], log_path)
    results.append({"endpoint": endpoint, "strategy": strategy, "model": model, "returncode": rc, "log": str(log_path)})

run_status = pd.DataFrame(results)
display(run_status)

if (run_status["returncode"] != 0).any():
    print("\n실패한 조합이 있습니다. 아래만 먼저 확인하세요:")
    display(run_status[run_status["returncode"] != 0])
else:
    print("모든 subprocess run이 0으로 종료되었습니다.")
```

```python

# 3) Aggregate latest runs if available
from pathlib import Path
all_summary = []
for p in sorted(RUNS_DIR.glob("subproc_allinone_*/summary_metrics.csv")):
    try:
        df = pd.read_csv(p)
        if not df.empty:
            df["summary_path"] = str(p)
            all_summary.append(df)
    except Exception:
        pass
if all_summary:
    summary = pd.concat(all_summary, ignore_index=True)
    summary = summary.sort_values(["endpoint", "mcc", "balanced_accuracy", "pr_auc"], ascending=[True, False, False, False])
    display(summary)
else:
    print("summary_metrics.csv가 아직 없습니다. 실패 로그부터 확인하세요.")
```

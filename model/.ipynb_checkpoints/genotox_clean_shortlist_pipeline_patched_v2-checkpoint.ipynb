# Genotox clean shortlist pipeline

이 노트북은 **스크립트 한 개를 단일 소스(source of truth)** 로 사용합니다.  
코드를 셀 안에 중복으로 쓰지 않고, `genotox_clean_shortlist_pipeline.py`를 import해서 실행합니다.

구성:
1. broadfp 깨끗하게 재생성/정리
2. shortlist 4개 실험만 재현
3. summary / best-by-endpoint / 기본 그림 확인

shortlist:
- Ames: `none + XGB`
- In vitro: `alert_bootstrap + XGB`
- In vitro benchmark: `compact logreg`
- In vivo: `alert_bootstrap + XGB`


이 버전은 `scaffold80_20` split 파일이 없을 때 `*_recommended_union_train/test.csv`도 자동 탐색합니다.

```
from pathlib import Path
import importlib.util
import sys
import pandas as pd

BASE_DIR = Path(r"C:\Users\Administrator\Documents\GitHub\Genotox_model\model")
SCRIPT_PATH = BASE_DIR / "genotox_clean_shortlist_pipeline.py"
if not SCRIPT_PATH.exists():
    SCRIPT_PATH = BASE_DIR / "genotox_clean_shortlist_pipeline_patched_v2.py"

# sandbox fallback
if not SCRIPT_PATH.exists():
    BASE_DIR = Path("/mnt/data/clean_shortlist_smoke2")
    SCRIPT_PATH = Path("/mnt/data/genotox_clean_shortlist_pipeline_patched_v2.py")

assert SCRIPT_PATH.exists(), f"스크립트를 찾지 못했습니다: {SCRIPT_PATH}"
print("BASE_DIR =", BASE_DIR)
print("SCRIPT_PATH =", SCRIPT_PATH)
```

```
spec = importlib.util.spec_from_file_location("genotox_clean_shortlist_pipeline", SCRIPT_PATH)
mod = importlib.util.module_from_spec(spec)
sys.modules["genotox_clean_shortlist_pipeline"] = mod
spec.loader.exec_module(mod)

print("PIPELINE_VERSION =", mod.PIPELINE_VERSION)
print(pd.DataFrame(mod.SHORTLIST_EXPERIMENTS)[["name","endpoint","strategy","model","blocks","fp_top_k"]])
```

```
# 1) broadfp를 먼저 깨끗하게 재생성/검증
# raw scaffold80_20 split이 없으면 recommended_union_train/test도 자동 탐색합니다.
FORCE_REBUILD_BROADFP = True

for ep in sorted({e["endpoint"] for e in mod.SHORTLIST_EXPERIMENTS}):
    train_path, test_path = mod.build_broadfp_for_endpoint(BASE_DIR, ep, force_rebuild=FORCE_REBUILD_BROADFP)
    print(ep, train_path.name, test_path.name)
```

```
# 2) shortlist 실행
SAVE_SHAP_BEST_ONLY = True
BASE_SEED = 42

summary = mod.run_shortlist(
    base_dir=BASE_DIR,
    force_rebuild_broadfp=False,   # 이미 위 셀에서 재생성했음
    save_shap_best_only=SAVE_SHAP_BEST_ONLY,
    base_seed=BASE_SEED,
)
summary
```

```
# 3) 저장 경로 및 요약
RUN_DIR = BASE_DIR / "runs" / "clean_shortlist_pipeline"
summary_path = RUN_DIR / "summary_metrics.csv"
best_path = RUN_DIR / "best_by_endpoint.csv"

print("RUN_DIR =", RUN_DIR)
print("summary_path exists =", summary_path.exists())
print("best_path exists =", best_path.exists())

summary_df = pd.read_csv(summary_path)
best_df = pd.read_csv(best_path)

display(summary_df.sort_values(["endpoint","mcc"], ascending=[True, False]))
display(best_df)
```

```
# 4) 대표 산출물 빠르게 보기
from IPython.display import Image, display

for exp_name in best_df["experiment_name"].tolist():
    exp_dir = RUN_DIR / exp_name
    print("\n==", exp_name, "==")
    for fname in ["confusion_matrix.png", "roc_curve.png", "pr_curve.png", "feature_importance_top.png"]:
        fpath = exp_dir / fname
        if fpath.exists():
            print(fname)
            display(Image(filename=str(fpath)))
```

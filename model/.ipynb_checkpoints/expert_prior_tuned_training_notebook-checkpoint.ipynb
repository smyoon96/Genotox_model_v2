# Endpoint-specific training with expert priors and scaffold-aware hyperparameter tuning

이 노트북은 unified train/test pipeline 출력 파일을 그대로 사용하고, endpoint별로 좁은 탐색 공간에서 하이퍼파라미터를 조절합니다.

```python

import os, sys, pathlib
BUNDLE_DIR = pathlib.Path("/mnt/data/expert_prior_tuned_training_code_bundle")
if str(BUNDLE_DIR) not in sys.path:
    sys.path.insert(0, str(BUNDLE_DIR))
import expert_prior_tuned_training as trainer
```

```python

ENDPOINT = "all"   # "ames", "invitro", "invivo", "all"
RUN_PLAN = "recommended"  # "recommended", "compact_only", "extended_only", "all"
OUTDIR = "/mnt/data/endpoint_model_runs_tuned_notebook"
USE_EXPERT_PRIOR = True
EXPERT_MODE = "auto"
TUNE_HYPERPARAMS = True
TUNE_N_ITER = 8
TUNING_CV_SPLITS = 5
SAVE_VISUALS = True
SAVE_MODELS = True
QM_FILE = None
QM_COLS = None
```

```python

summary = trainer.run_experiments(
    endpoint=ENDPOINT,
    run_plan=RUN_PLAN,
    outdir=OUTDIR,
    save_models=SAVE_MODELS,
    qm_file=QM_FILE,
    qm_cols_arg=QM_COLS,
    save_visuals=SAVE_VISUALS,
    use_expert_prior=USE_EXPERT_PRIOR,
    expert_mode=EXPERT_MODE,
    tune_hyperparams=TUNE_HYPERPARAMS,
    tune_n_iter=TUNE_N_ITER,
    tuning_cv_splits=TUNING_CV_SPLITS,
)
summary
```

```python

import pandas as pd, json, os
pd.read_csv(os.path.join(OUTDIR, "summary_metrics.csv"))
```

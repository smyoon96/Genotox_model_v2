# Expert prior weighted endpoint training

이 노트북은 unified train/test pipeline 출력 파일을 그대로 사용합니다.

기능:
- endpoint-specific compact / extended 모델 학습
- SHAP 시각화 및 혼동행렬 저장
- high-confidence positive alert를 반영한 `expert_alert_score` 추가
- 필요 시 `sample_weight multiplier` 적용


```python
from pathlib import Path
import json
import sys

NOTEBOOK_DIR = Path.cwd()
if str(NOTEBOOK_DIR) not in sys.path:
    sys.path.insert(0, str(NOTEBOOK_DIR))

from expert_prior_weighted_training import run_experiments, load_expert_config

CONFIG_PATH = NOTEBOOK_DIR / "expert_prior_config.json"
OUTDIR = NOTEBOOK_DIR / "runs_expert_prior_notebook"

ENDPOINT = "all"          # "ames" / "invitro" / "invivo" / "all"
RUN_PLAN = "recommended"  # "recommended" / "compact_only" / "extended_only" / "all"
USE_EXPERT_PRIOR = True
EXPERT_MODE = "auto"      # "auto" / "none" / "feature" / "sample_weight" / "both"
USE_QM = False
QM_FILE = None
MERGE_KEY = "No"
SAVE_MODELS = True
SAVE_VISUALS = True

OUTDIR.mkdir(parents=True, exist_ok=True)
print("Notebook dir:", NOTEBOOK_DIR)
print("Config path:", CONFIG_PATH)
print("Output dir:", OUTDIR)
```

```python
expert_cfg = load_expert_config(str(CONFIG_PATH))
expert_cfg
```

## 학습 실행

```python
results_df = run_experiments(
    endpoint=ENDPOINT,
    run_plan=RUN_PLAN,
    outdir=str(OUTDIR),
    save_models=SAVE_MODELS,
    qm_file=QM_FILE if USE_QM else None,
    merge_key=MERGE_KEY,
    save_visuals=SAVE_VISUALS,
    use_expert_prior=USE_EXPERT_PRIOR,
    expert_mode=EXPERT_MODE,
    expert_config_path=str(CONFIG_PATH),
)
results_df
```

## 저장 결과 확인

```python
summary_path = OUTDIR / "summary_metrics.csv"
print(summary_path)
results_df.sort_values(["endpoint", "tier", "model_name"]).reset_index(drop=True)
```

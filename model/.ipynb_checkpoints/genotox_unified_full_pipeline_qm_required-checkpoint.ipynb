# Genotox unified full pipeline (QM required)

이 노트북은 아래 6단계를 **하나의 파이프라인**으로 연결합니다.

1. raw 데이터 로드 및 데이터 버전 구분  
2. canonical / standardized / preprocessed feature-ready 테이블 생성  
3. scaffold-aware train/test split  
4. 작용기, 물리화학적 특성, **QM(qm_descriptors.py 기반 필수)**, fingerprint 데이터세트 구축  
5. 언더샘플링(훈련 폴드 내부) + 모델 구축  
6. 혼동행렬, ROC, PR, threshold sweep, feature importance, SHAP(옵션) 시각화

기본 스크립트는 같은 폴더의 `genotox_unified_full_pipeline_qm_required.py` 를 사용합니다.  
또한 같은 폴더 또는 `data/` 폴더에 `qm_descriptors.py` 가 있어야 합니다.


```python
from pathlib import Path
import importlib.util
import sys
import pandas as pd

# 로컬 우선, 없으면 현재 폴더
BASE_DIR = Path(r"C:\Users\Administrator\Documents\GitHub\Genotox_model\model")
if not BASE_DIR.exists():
    BASE_DIR = Path.cwd()

SCRIPT_PATH = BASE_DIR / "genotox_unified_full_pipeline_qm_required.py"
if not SCRIPT_PATH.exists():
    raise FileNotFoundError(f"pipeline script not found: {SCRIPT_PATH}")

spec = importlib.util.spec_from_file_location("genotox_pipeline", SCRIPT_PATH)
mod = importlib.util.module_from_spec(spec)
sys.modules["genotox_pipeline"] = mod
spec.loader.exec_module(mod)

print("BASE_DIR:", BASE_DIR)
print("SCRIPT_PATH:", SCRIPT_PATH)
print("ENDPOINTS:", list(mod.ENDPOINTS.keys()))
```

```python
# =========================
# 사용자 설정
# =========================
OUTPUT_ROOT = BASE_DIR / "runs" / "unified_full_pipeline"
ENDPOINTS = ["ames", "invitro", "invivo"]

# 기본: canonical 기반
REPRESENTATION_MAP = {
    "ames": "canonical",
    "invitro": "canonical",
    "invivo": "canonical",
}

# endpoint별 기본 sampling 전략
SAMPLING_MAP = {
    "ames": "none",
    "invitro": "coverage_hardneg",
    "invivo": "coverage_hardneg",
}

# endpoint별 기본 모델
MODEL_MAP = {
    "ames": "xgb",
    "invitro": "xgb",
    "invivo": "xgb",
}

# feature block 구성
FEATURE_BLOCKS_MAP = {
    "ames": ["physchem", "rules", "fg_present", "fg_count", "fingerprint", "qm", "expert", "scaffold_meta"],
    "invitro": ["physchem", "rules", "fg_present", "fg_count", "fingerprint", "qm", "expert", "scaffold_meta"],
    "invivo": ["physchem", "rules", "fg_present", "fg_count", "fingerprint", "qm", "expert", "scaffold_meta"],
}

# QM 파일이 있으면 넣고, 없으면 None
QM_FILE = None
# QM_FILE = BASE_DIR / "data" / "qm_descriptors.csv"

# xTB가 설치되어 있을 때만 True 권장
COMPUTE_QM_IF_MISSING = False

FP_BITS = 256
FP_TOP_K_MAP = {
    "ames": 128,
    "invitro": 96,
    "invivo": 64,
}
TUNE_ITER_MAP = {
    "ames": 16,
    "invitro": 12,
    "invivo": 10,
}
CV_SPLITS = 3
SAVE_SHAP_BEST_ONLY = True
FORCE_REBUILD_FEATURES = True
BASE_SEED = 42

print("OUTPUT_ROOT:", OUTPUT_ROOT)
```

```python
# =========================
# 1~4 단계: raw -> standardized -> preprocessed -> split -> feature build (+ required QM)
# =========================
prepared = {}
for ep in ENDPOINTS:
    prepared[ep] = mod.prepare_endpoint_dataset(
        base_dir=BASE_DIR,
        output_root=OUTPUT_ROOT,
        endpoint=ep,
        representation=REPRESENTATION_MAP[ep],
        fp_bits=FP_BITS,
        qm_file=QM_FILE,
        compute_qm_if_missing=COMPUTE_QM_IF_MISSING,
        force_rebuild=FORCE_REBUILD_FEATURES,
    )
    train_df = prepared[ep]["train_df"]
    test_df = prepared[ep]["test_df"]
    print(ep, "train:", train_df.shape, "test:", test_df.shape)
    print("  train pos rate:", round(train_df["label"].mean(), 4), "test pos rate:", round(test_df["label"].mean(), 4))
    print("  blocks:", {k: len(v) for k, v in mod.detect_feature_blocks(train_df).items()})
```

```python
# =========================
# 5~6 단계: sampling + model + visualization
# =========================
summary = mod.run_full_pipeline(
    base_dir=BASE_DIR,
    output_root=OUTPUT_ROOT,
    endpoints=ENDPOINTS,
    representation_map=REPRESENTATION_MAP,
    feature_blocks_map=FEATURE_BLOCKS_MAP,
    sampling_map=SAMPLING_MAP,
    model_map=MODEL_MAP,
    qm_file=QM_FILE,
    compute_qm_if_missing=COMPUTE_QM_IF_MISSING,
    fp_bits=FP_BITS,
    fp_top_k_map=FP_TOP_K_MAP,
    tune_iter_map=TUNE_ITER_MAP,
    cv_splits=CV_SPLITS,
    save_shap_best_only=SAVE_SHAP_BEST_ONLY,
    force_rebuild_features=False,
    base_seed=BASE_SEED,
)
summary
```

```python
# summary 파일 확인
summary_path = OUTPUT_ROOT / "_summary" / "summary_metrics.csv"
best_path = OUTPUT_ROOT / "_summary" / "best_by_endpoint.csv"
dashboard_path = OUTPUT_ROOT / "_summary" / "summary_dashboard.png"

print("summary_path:", summary_path)
print("best_path:", best_path)
print("dashboard_path:", dashboard_path)

summary_df = pd.read_csv(summary_path) if summary_path.exists() else pd.DataFrame()
best_df = pd.read_csv(best_path) if best_path.exists() else pd.DataFrame()

display(summary_df.sort_values(["endpoint", "mcc"], ascending=[True, False]) if not summary_df.empty else summary_df)
display(best_df)
```

# Genotox subprocess pipeline (patched)

이 노트북은 실행 로직을 문자열로 내장하지 않고, 외부 `.py` 스크립트를 `_notebook_runtime/`로 복사해서 실행합니다.

반영된 패치:
- CV 내부 fingerprint bit 선택으로 inner-CV leakage 완화
- notebook/script 단일 소스화 (`shutil.copy2`)
- randomized search (`ParameterSampler`)
- `SeedSequence` 기반 시드 생성
- `is_categorical_dtype` 제거
- 문자열/범주형 컬럼 강제 분리 및 numeric coercion
- `trained_bundle.joblib` 경량화
- blanket warning ignore 제거
- RDKit `ConvertToNumpyArray` 사용


```python
from pathlib import Path
import json
import shutil
import subprocess
import sys
import pandas as pd

BASE_DIR = Path(r"C:\Users\Administrator\Documents\GitHub\Genotox_model\model")
DATA_DIR = BASE_DIR / "data"
RUNS_DIR = BASE_DIR / "runs"
RUNTIME_DIR = BASE_DIR / "_notebook_runtime"
RUNTIME_DIR.mkdir(parents=True, exist_ok=True)

# 외부 파이썬 스크립트 원본 경로
# 1순위: model 폴더 내 이미 배치된 patched 스크립트
# 2순위: 노트북과 같은 폴더
SCRIPT_CANDIDATES = [
    BASE_DIR / "genotox_subprocess_pipeline_patched_v2.py",
    Path.cwd() / "genotox_subprocess_pipeline_patched_v2.py",
]
SCRIPT_SRC = next((p for p in SCRIPT_CANDIDATES if p.exists()), None)
if SCRIPT_SRC is None:
    raise FileNotFoundError("genotox_subprocess_pipeline_patched_v2.py 파일을 찾지 못했습니다. model 폴더 또는 현재 폴더에 두세요.")

SCRIPT_DST = RUNTIME_DIR / "genotox_subprocess_pipeline.py"
shutil.copy2(SCRIPT_SRC, SCRIPT_DST)

CONFIG = {
    "base_dir": str(BASE_DIR),
    "data_dir": str(DATA_DIR),
    "runs_dir": str(RUNS_DIR),
    "endpoints": ["ames", "invitro", "invivo"],
    "strategies": ["none", "alert_bootstrap"],  # 시작은 보수적으로
    "models": ["logreg", "xgb"],
    "feature_blocks": ["physchem", "rules", "fg_present", "fg_count", "fingerprint", "qm"],
    "fp_bits": 256,
    "fp_top_k": 128,
    "qm_file": None,
    "qm_merge_key": "No",
    "cv_splits": 3,
    "random_state": 42,
    "tuning_metric": "mcc",
    "search_n_iter": 16,
    "search_n_iter_logreg": 16,
    "search_n_iter_xgb": 12,
    "shap_background_n": 200,
    "shap_explain_n": 300,
    "force_rebuild_broadfp": True,
}
CONFIG_PATH = RUNTIME_DIR / "pipeline_config.json"
CONFIG_PATH.write_text(json.dumps(CONFIG, ensure_ascii=False, indent=2), encoding="utf-8")

print("SCRIPT_SRC:", SCRIPT_SRC)
print("SCRIPT_DST:", SCRIPT_DST)
print("CONFIG_PATH:", CONFIG_PATH)
print(json.dumps(CONFIG, ensure_ascii=False, indent=2))
```

```python
# Build step
LOG_PATH = RUNTIME_DIR / "build_subprocess.log"
cmd = [sys.executable, str(SCRIPT_DST), "--config-json", str(CONFIG_PATH), "--mode", "build"]
print("Running build:", " ".join(cmd))
result = subprocess.run(cmd, text=True, capture_output=True)
LOG_PATH.write_text(
    "STDOUT\n" + result.stdout + "\n\nSTDERR\n" + result.stderr,
    encoding="utf-8"
)
print("returncode =", result.returncode)
stdout_tail = "\n".join(result.stdout.splitlines()[-40:])
stderr_tail = "\n".join(result.stderr.splitlines()[-80:])
if stdout_tail:
    print("---- stdout tail ----")
    print(stdout_tail)
if stderr_tail:
    print("---- stderr tail ----")
    print(stderr_tail)
print("log file:", LOG_PATH)
if result.returncode != 0:
    raise RuntimeError("Build step failed. See log file above.")
```

```python
# Run step
LOG_PATH = RUNTIME_DIR / "run_subprocess.log"
cmd = [sys.executable, str(SCRIPT_DST), "--config-json", str(CONFIG_PATH), "--mode", "run"]
print("Running run:", " ".join(cmd))
result = subprocess.run(cmd, text=True, capture_output=True)
LOG_PATH.write_text(
    "STDOUT\n" + result.stdout + "\n\nSTDERR\n" + result.stderr,
    encoding="utf-8"
)
print("returncode =", result.returncode)
stdout_tail = "\n".join(result.stdout.splitlines()[-40:])
stderr_tail = "\n".join(result.stderr.splitlines()[-80:])
if stdout_tail:
    print("---- stdout tail ----")
    print(stdout_tail)
if stderr_tail:
    print("---- stderr tail ----")
    print(stderr_tail)
print("log file:", LOG_PATH)
if result.returncode != 0:
    raise RuntimeError("Run step failed. See log file above.")
```

```python
# SHAP step
LOG_PATH = RUNTIME_DIR / "shap_subprocess.log"
cmd = [sys.executable, str(SCRIPT_DST), "--config-json", str(CONFIG_PATH), "--mode", "shap"]
print("Running shap:", " ".join(cmd))
result = subprocess.run(cmd, text=True, capture_output=True)
LOG_PATH.write_text(
    "STDOUT\n" + result.stdout + "\n\nSTDERR\n" + result.stderr,
    encoding="utf-8"
)
print("returncode =", result.returncode)
stdout_tail = "\n".join(result.stdout.splitlines()[-40:])
stderr_tail = "\n".join(result.stderr.splitlines()[-80:])
if stdout_tail:
    print("---- stdout tail ----")
    print(stdout_tail)
if stderr_tail:
    print("---- stderr tail ----")
    print(stderr_tail)
print("log file:", LOG_PATH)
if result.returncode != 0:
    print("SHAP step failed or was skipped. See log file above.")
```

```python
# 결과 요약 읽기
latest_path = RUNS_DIR / "LATEST_SUBPROC_RUN_PATH.txt"
if latest_path.exists():
    RUN_ROOT = Path(latest_path.read_text(encoding="utf-8").strip())
else:
    RUN_ROOT = None

print("RUN_ROOT:", RUN_ROOT)

summary_path = RUN_ROOT / "summary_metrics.csv" if RUN_ROOT else None
best_path = RUN_ROOT / "best_by_endpoint.csv" if RUN_ROOT else None

summary_df = pd.read_csv(summary_path) if summary_path and summary_path.exists() and summary_path.stat().st_size > 1 else pd.DataFrame()
best_df = pd.read_csv(best_path) if best_path and best_path.exists() and best_path.stat().st_size > 1 else pd.DataFrame()

print("summary exists:", bool(summary_path and summary_path.exists()), "| rows:", len(summary_df))
print("best exists:", bool(best_path and best_path.exists()), "| rows:", len(best_df))

if not summary_df.empty and {"endpoint", "mcc"}.issubset(summary_df.columns):
    display(summary_df.sort_values(["endpoint", "mcc"], ascending=[True, False]))
else:
    print("summary_metrics.csv가 비어 있거나 핵심 컬럼이 없습니다.")

if not best_df.empty:
    display(best_df)

for name in ["summary_mcc.png", "summary_balanced_accuracy.png", "summary_pr_auc.png"]:
    p = RUN_ROOT / name if RUN_ROOT else None
    if p and p.exists():
        print(p)
```

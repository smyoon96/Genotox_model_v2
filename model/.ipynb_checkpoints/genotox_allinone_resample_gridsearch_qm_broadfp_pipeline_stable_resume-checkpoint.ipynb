
# Genotox all-in-one training pipeline

- broadfp 데이터셋 자동 탐색 및 필요 시 생성(옵션)
- feature block 선택  
  - physicochemical

  - fingerprint

  - baseline

  - hybrid (alert-guided bootstrap + SMOTENC)
- threshold tuning
- confusion matrix / ROC / PR / threshold sweep

```python

# =========================
# 3. OPTIONAL: BROADFP BUILD HELPERS
# =========================
def find_first_existing(patterns: List[str], root: Path) -> Optional[Path]:
    for p in patterns:
        candidate = root / p
        if candidate.exists():
            return candidate
    return None

def locate_fg_analysis_source() -> Optional[Path]:
    for p in FG_ANALYSIS_CANDIDATES:
        if p.exists():
            return p
    return None

def read_csv_from_zip_or_dir(source: Path, relative_path: str) -> pd.DataFrame:
    if source.is_dir():
        return pd.read_csv(source / relative_path)
    with zipfile.ZipFile(source) as zf:
        with zf.open(relative_path) as f:
            return pd.read_csv(f)

def build_rules(df: pd.DataFrame, endpoint: str) -> pd.DataFrame:
    out = df.copy()
    for rule_name, (src_col, op, thr) in ENDPOINT_RULE_BUILDERS[endpoint].items():
        if src_col not in out.columns:
            out[rule_name] = 0
            continue
        if op == ">=":
            out[rule_name] = (out[src_col].fillna(0) >= thr).astype(int)
        elif op == "==":
            out[rule_name] = (out[src_col].fillna(0) == thr).astype(int)
        else:
            raise ValueError(f"Unsupported op: {op}")
    return out

def compute_morgan_fp(smiles_series: pd.Series, n_bits: int = 256, radius: int = 2) -> pd.DataFrame:
    try:
        from rdkit import Chem
        from rdkit.Chem import AllChem
    except Exception as e:
        raise ImportError(
            "broadfp 자동 생성을 위해 RDKit이 필요합니다. "
            "conda install -c conda-forge rdkit 로 설치하세요."
        ) from e

    fps = np.zeros((len(smiles_series), n_bits), dtype=np.int8)
    for i, smi in enumerate(smiles_series.fillna("").astype(str)):
        mol = Chem.MolFromSmiles(smi)
        if mol is None:
            continue
        bitvect = AllChem.GetMorganFingerprintAsBitVect(mol, radius, nBits=n_bits)
        arr = np.zeros((n_bits,), dtype=np.int8)
        # RDKit -> numpy
        from rdkit import DataStructs
        DataStructs.ConvertToNumpyArray(bitvect, arr)
        fps[i, :] = arr
    return pd.DataFrame(fps, columns=[f"fp_morgan2_{i}" for i in range(n_bits)])

def infer_scaffold_group_type(df: pd.DataFrame) -> pd.Series:
    if "scaffold_group_type" in df.columns:
        return df["scaffold_group_type"].fillna("murcko_scaffold")
    scaffold = df.get("murcko_scaffold", pd.Series("", index=df.index)).fillna("")
    return np.where(scaffold.astype(str).str.len() > 0, "murcko_scaffold", "acyclic_anonymous_graph")

def load_split_base(endpoint: str, split_kind: str) -> pd.DataFrame:
    patterns = SOURCE_SPLIT_PATTERNS[endpoint][split_kind]
    path = find_first_existing(patterns, DATA_DIR)
    if path is None:
        raise FileNotFoundError(f"{endpoint}/{split_kind} source split 파일을 찾지 못했습니다: {patterns}")
    return pd.read_csv(path)

def build_broadfp_for_endpoint(endpoint: str) -> Tuple[pd.DataFrame, pd.DataFrame]:
    fg_source = locate_fg_analysis_source()
    if fg_source is None:
        raise FileNotFoundError(
            "fg_descriptor_analysis.zip 또는 fg_descriptor_analysis 폴더를 찾지 못했습니다.\n"
            "model/data 혹은 model 폴더 아래에 두세요."
        )

    analysis_stem = ANALYSIS_ENDPOINT_STEMS[endpoint]
    canonical_rel = f"fg_descriptor_analysis/{analysis_stem}/{analysis_stem}_canonical_analysis.csv"
    views_rel = f"fg_descriptor_analysis/{analysis_stem}/{analysis_stem}_preprocessed_views.csv"

    canonical = read_csv_from_zip_or_dir(fg_source, canonical_rel)
    views = read_csv_from_zip_or_dir(fg_source, views_rel)

    canonical = canonical[canonical["valid"].fillna(False)].copy()
    canonical = canonical.drop(columns=[c for c in ["representation", "valid", "analysis_smiles", "metal_elements"] if c in canonical.columns], errors="ignore")

    views_keep = [c for c in ["No", "SMILES", "canonical_smiles", "standardized_smiles"] if c in views.columns]
    views = views[views_keep].copy()

    train_base = load_split_base(endpoint, "train")
    test_base = load_split_base(endpoint, "test")

    def enrich(base_df: pd.DataFrame) -> pd.DataFrame:
        meta_cols = [c for c in base_df.columns if c in ["No","SMILES","label","murcko_scaffold","scaffold_group","scaffold_group_type","scaffold_split_80_20"]]
        meta = base_df[meta_cols].copy()

        merged = meta.merge(canonical, on=["No", "label"], how="left")
        if "SMILES" not in merged.columns or merged["SMILES"].isna().all():
            merged = merged.drop(columns=["SMILES"], errors="ignore").merge(views[["No","SMILES"]], on="No", how="left")
        else:
            merged = merged.merge(views[["No","SMILES"]], on="No", how="left", suffixes=("", "_views"))
            if "SMILES_views" in merged.columns:
                merged["SMILES"] = merged["SMILES"].fillna(merged["SMILES_views"])
                merged = merged.drop(columns=["SMILES_views"])

        if "murcko_scaffold" not in merged.columns:
            merged["murcko_scaffold"] = ""
        if "scaffold_group" not in merged.columns:
            merged["scaffold_group"] = merged["murcko_scaffold"].fillna("").replace("", np.nan)
            missing = merged["scaffold_group"].isna()
            merged.loc[missing, "scaffold_group"] = "acyclic::" + merged.loc[missing, "No"].astype(str)
        merged["scaffold_group_type"] = infer_scaffold_group_type(merged)

        merged = build_rules(merged, endpoint)

        fps = compute_morgan_fp(merged["SMILES"])
        merged = pd.concat([merged.reset_index(drop=True), fps.reset_index(drop=True)], axis=1)

        # broadfp와 유사하게 정리
        # 중복 컬럼 제거
        merged = merged.loc[:, ~merged.columns.duplicated()]
        return merged

    train_broad = enrich(train_base)
    test_broad = enrich(test_base)

    train_out = DATA_DIR / f"{ENDPOINT_FILE_STEMS[endpoint]}_broadfp_train.csv"
    test_out = DATA_DIR / f"{ENDPOINT_FILE_STEMS[endpoint]}_broadfp_test.csv"
    train_broad.to_csv(train_out, index=False)
    test_broad.to_csv(test_out, index=False)
    print(f"[AUTO-BUILD] saved -> {train_out}")
    print(f"[AUTO-BUILD] saved -> {test_out}")
    return train_broad, test_broad

def ensure_broadfp_files(endpoint: str) -> Tuple[Path, Path]:
    train_path = DATA_DIR / f"{ENDPOINT_FILE_STEMS[endpoint]}_broadfp_train.csv"
    test_path = DATA_DIR / f"{ENDPOINT_FILE_STEMS[endpoint]}_broadfp_test.csv"
    if train_path.exists() and test_path.exists():
        return train_path, test_path
    if not AUTO_BUILD_BROADFP_IF_MISSING:
        raise FileNotFoundError(
            f"broadfp 파일이 없습니다: {train_path.name}, {test_path.name}\n"
            "AUTO_BUILD_BROADFP_IF_MISSING=True 로 바꾸거나 파일을 먼저 생성하세요."
        )
    print(f"[AUTO-BUILD] broadfp files missing for {endpoint}. Building...")
    build_broadfp_for_endpoint(endpoint)
    return train_path, test_path
```

```python

# =========================
# 4. DATA LOAD / QM MERGE / FEATURE BLOCK SELECTION
# =========================
def detect_qm_file() -> Optional[Path]:
    if QM_FILE is not None:
        p = Path(QM_FILE)
        return p if p.exists() else None

    candidates = []
    patterns = ["*qm*.csv", "*quantum*.csv", "*descriptor*qm*.csv"]
    for pat in patterns:
        candidates.extend(sorted(DATA_DIR.glob(pat)))
        candidates.extend(sorted(BASE_DIR.glob(pat)))
    return candidates[0] if candidates else None

def merge_qm_if_available(train_df: pd.DataFrame, test_df: pd.DataFrame) -> Tuple[pd.DataFrame, pd.DataFrame, List[str], Optional[Path]]:
    qm_path = detect_qm_file()
    if qm_path is None:
        return train_df, test_df, [], None

    qm_df = pd.read_csv(qm_path)
    merge_key = QM_MERGE_KEY if QM_MERGE_KEY in qm_df.columns else None
    if merge_key is None:
        for fallback in ["No", "SMILES", "canonical_smiles", "smiles"]:
            if fallback in qm_df.columns and fallback in train_df.columns:
                merge_key = fallback
                break
    if merge_key is None:
        print(f"[QM] merge key not found in {qm_path.name}; skipping QM.")
        return train_df, test_df, [], qm_path

    existing_cols = set(train_df.columns)
    qm_cols = [c for c in qm_df.columns if c != merge_key and c not in existing_cols]

    train_df = train_df.merge(qm_df[[merge_key] + qm_cols], on=merge_key, how="left")
    test_df = test_df.merge(qm_df[[merge_key] + qm_cols], on=merge_key, how="left")
    print(f"[QM] merged {len(qm_cols)} columns from {qm_path.name} using key '{merge_key}'")
    return train_df, test_df, qm_cols, qm_path

def add_alert_score_features(df: pd.DataFrame, endpoint: str) -> pd.DataFrame:
    df = df.copy()
    pos_cols = [c for c in ENDPOINT_POS_ALERTS.get(endpoint, []) if c in df.columns]
    neg_cols = [c for c in ENDPOINT_NEG_ALERTS.get(endpoint, []) if c in df.columns]
    df["expert_positive_alert_score"] = df[pos_cols].fillna(0).sum(axis=1) if pos_cols else 0.0
    df["expert_negative_alert_score"] = df[neg_cols].fillna(0).sum(axis=1) if neg_cols else 0.0
    return df

def choose_feature_columns(train_df: pd.DataFrame, qm_cols: List[str]) -> Tuple[List[str], FeatureAudit]:
    physchem_cols = [c for c in PHYSCHEM_COLS if c in train_df.columns] if FEATURE_BLOCKS["physchem"] else []
    rule_cols = [c for c in train_df.columns if c.startswith("rule_")] if FEATURE_BLOCKS["rules"] else []
    fg_present_cols = [c for c in train_df.columns if c.startswith("fg_") and c.endswith("_present")] if FEATURE_BLOCKS["fg_present"] else []
    fg_count_cols = [c for c in train_df.columns if c.startswith("fg_") and c.endswith("_count")] if FEATURE_BLOCKS["fg_count"] else []
    fp_cols = [c for c in train_df.columns if c.startswith("fp_morgan2_")] if FEATURE_BLOCKS["fingerprint"] else []
    qm_use_cols = [c for c in qm_cols if c in train_df.columns] if FEATURE_BLOCKS["qm"] else []
    extra_cols = [c for c in ["expert_positive_alert_score", "expert_negative_alert_score"] if c in train_df.columns]

    feature_cols = physchem_cols + rule_cols + fg_present_cols + fg_count_cols + fp_cols + qm_use_cols + TEXT_CATEGORICAL_COLS + extra_cols
    feature_cols = [c for c in feature_cols if c in train_df.columns and c not in META_COLS]

    # fingerprint bit selection
    if FEATURE_BLOCKS["fingerprint"] and FP_TOP_K is not None and len(fp_cols) > FP_TOP_K:
        fp_sub = train_df[fp_cols].fillna(0)
        y = train_df[TARGET_COL].astype(int)
        scores, _ = chi2(fp_sub, y)
        order = np.argsort(-scores)
        keep_fp = [fp_cols[i] for i in order[:FP_TOP_K]]
        feature_cols = [c for c in feature_cols if c not in fp_cols] + keep_fp
        fp_cols = keep_fp

    audit = FeatureAudit(
        endpoint="",
        n_total_features=len(feature_cols),
        n_physchem=len(physchem_cols),
        n_rules=len(rule_cols),
        n_fg_present=len(fg_present_cols),
        n_fg_count=len(fg_count_cols),
        n_fp=len(fp_cols),
        n_qm=len(qm_use_cols),
        selected_example=feature_cols[:20],
    )
    return feature_cols, audit

def load_endpoint_data(endpoint: str) -> Tuple[pd.DataFrame, pd.DataFrame, List[str], FeatureAudit]:
    train_path, test_path = ensure_broadfp_files(endpoint)
    train_df = pd.read_csv(train_path)
    test_df = pd.read_csv(test_path)

    train_df = add_alert_score_features(train_df, endpoint)
    test_df = add_alert_score_features(test_df, endpoint)

    train_df, test_df, qm_cols, qm_path = merge_qm_if_available(train_df, test_df)
    feature_cols, audit = choose_feature_columns(train_df, qm_cols)
    audit.endpoint = endpoint

    print(f"[LOAD] {endpoint}: train={train_df.shape}, test={test_df.shape}")
    print(f"[LOAD] qm_path={qm_path}")
    print(f"[AUDIT] total={audit.n_total_features} | physchem={audit.n_physchem} | rules={audit.n_rules} | fg_present={audit.n_fg_present} | fg_count={audit.n_fg_count} | fp={audit.n_fp} | qm={audit.n_qm}")
    print(f"[AUDIT] feature examples: {audit.selected_example[:12]}")
    return train_df, test_df, feature_cols, audit

feature_audits = []
loaded = {}
for ep in ENDPOINTS:
    train_df, test_df, feature_cols, audit = load_endpoint_data(ep)
    feature_audits.append(audit.__dict__)
    loaded[ep] = {"train": train_df, "test": test_df, "feature_cols": feature_cols}

feature_audit_df = pd.DataFrame(feature_audits)
display(feature_audit_df)
```

```python

# =========================
# 5. MODELING / RESAMPLING UTILITIES
# =========================
def detect_binary_like_cols(df: pd.DataFrame, cols: List[str]) -> List[str]:
    out = []
    for c in cols:
        if c in TEXT_CATEGORICAL_COLS:
            continue
        s = df[c]
        if pd.api.types.is_bool_dtype(s):
            out.append(c)
            continue
        if pd.api.types.is_numeric_dtype(s):
            vals = pd.Series(s.dropna().unique())
            if len(vals) > 0 and set(np.unique(vals.astype(float))).issubset({0.0, 1.0}):
                out.append(c)
    return out

class AlertGuidedBootstrapSampler(BaseEstimator):
    """
    고확신 positive/negative alert family 내부에서만 bootstrap duplication을 수행.
    완전히 새로운 분자를 만드는 것이 아니라, training fold 내부의 high-confidence sample을 재표본화하는 방식.
    """
    def __init__(self, endpoint: str, pos_multiplier: float = 1.50, neg_multiplier: float = 1.10, random_state: int = 42):
        self.endpoint = endpoint
        self.pos_multiplier = pos_multiplier
        self.neg_multiplier = neg_multiplier
        self.random_state = random_state

    def fit_resample(self, X, y):
        if not isinstance(X, pd.DataFrame):
            X = pd.DataFrame(X)
        X = X.reset_index(drop=True)
        y = pd.Series(y).reset_index(drop=True)

        rng = np.random.default_rng(self.random_state)
        pos_cols = [c for c in ENDPOINT_POS_ALERTS.get(self.endpoint, []) if c in X.columns]
        neg_cols = [c for c in ENDPOINT_NEG_ALERTS.get(self.endpoint, []) if c in X.columns]

        pos_scores = X[pos_cols].fillna(0).sum(axis=1) if pos_cols else pd.Series(0, index=X.index)
        neg_scores = X[neg_cols].fillna(0).sum(axis=1) if neg_cols else pd.Series(0, index=X.index)

        pos_idx = X.index[(y == 1) & (pos_scores > 0)].to_numpy()
        neg_idx = X.index[(y == 0) & (neg_scores > 0)].to_numpy()

        extra_idx = []

        if len(pos_idx) and self.pos_multiplier > 1:
            n_extra = int(round((self.pos_multiplier - 1.0) * len(pos_idx)))
            p = pos_scores.loc[pos_idx].to_numpy(dtype=float)
            p = p / p.sum() if p.sum() > 0 else None
            extra_idx.extend(rng.choice(pos_idx, size=n_extra, replace=True, p=p).tolist())

        if len(neg_idx) and self.neg_multiplier > 1:
            n_extra = int(round((self.neg_multiplier - 1.0) * len(neg_idx)))
            p = neg_scores.loc[neg_idx].to_numpy(dtype=float)
            p = p / p.sum() if p.sum() > 0 else None
            extra_idx.extend(rng.choice(neg_idx, size=n_extra, replace=True, p=p).tolist())

        if not extra_idx:
            return X, y.to_numpy()

        X_res = pd.concat([X, X.loc[extra_idx]], axis=0, ignore_index=True)
        y_res = pd.concat([y, y.loc[extra_idx]], axis=0, ignore_index=True).to_numpy()
        return X_res, y_res

def build_preprocessor(X_train: pd.DataFrame, feature_cols: List[str], model_name: str):
    categorical_cols = [c for c in feature_cols if c in TEXT_CATEGORICAL_COLS]
    numeric_cols = [c for c in feature_cols if c not in categorical_cols]

    num_steps = [("imputer", SimpleImputer(strategy="median"))]
    if model_name == "logreg":
        num_steps.append(("scaler", StandardScaler(with_mean=False)))

    cat_pipe = Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("ohe", OneHotEncoder(handle_unknown="ignore")),
    ])

    preprocessor = ColumnTransformer([
        ("num", Pipeline(num_steps), numeric_cols),
        ("cat", cat_pipe, categorical_cols),
    ], remainder="drop", sparse_threshold=0.3)

    return preprocessor, numeric_cols, categorical_cols

def build_model(model_name: str, endpoint: str):
    if model_name == "logreg":
        return LogisticRegression(
            solver="saga",
            penalty="elasticnet",
            max_iter=5000,
            random_state=RANDOM_STATE,
            n_jobs=None,
        )
    if model_name == "xgb":
        return XGBClassifier(
            objective="binary:logistic",
            eval_metric="logloss",
            tree_method="hist",
            n_jobs=1,
            random_state=RANDOM_STATE,
            verbosity=0,
        )
    raise ValueError(f"Unknown model: {model_name}")

def get_param_grid(model_name: str, endpoint: str):
    cfg = ENDPOINT_MODEL_CFG[endpoint]
    return cfg["logreg_grid"] if model_name == "logreg" else cfg["xgb_grid"]

def build_sampler(strategy: str, X_train: pd.DataFrame, feature_cols: List[str], endpoint: str):
    binary_like_cols = detect_binary_like_cols(X_train, feature_cols)
    smotenc_cat_cols = [c for c in feature_cols if c in TEXT_CATEGORICAL_COLS] + [c for c in binary_like_cols if c in feature_cols]

    if strategy == "none":
        return None
    if strategy == "alert_bootstrap":
        return AlertGuidedBootstrapSampler(endpoint=endpoint, random_state=RANDOM_STATE)
    if strategy == "smotenc":
        return SMOTENC(
            categorical_features=smotenc_cat_cols,
            k_neighbors=3,
            sampling_strategy="auto",
            random_state=RANDOM_STATE,
        )
    if strategy == "hybrid":
        return [
            ("alert_bootstrap", AlertGuidedBootstrapSampler(endpoint=endpoint, random_state=RANDOM_STATE)),
            ("smotenc", SMOTENC(
                categorical_features=smotenc_cat_cols,
                k_neighbors=3,
                sampling_strategy="auto",
                random_state=RANDOM_STATE,
            )),
        ]
    raise ValueError(f"Unknown resampling strategy: {strategy}")

def build_pipeline(X_train: pd.DataFrame, feature_cols: List[str], endpoint: str, model_name: str, strategy: str):
    preprocessor, _, _ = build_preprocessor(X_train, feature_cols, model_name)
    model = build_model(model_name, endpoint)
    sampler = build_sampler(strategy, X_train, feature_cols, endpoint)

    if sampler is None:
        steps = [("preprocessor", preprocessor), ("model", model)]
    elif isinstance(sampler, list):
        steps = sampler + [("preprocessor", preprocessor), ("model", model)]
    else:
        steps = [("sampler", sampler), ("preprocessor", preprocessor), ("model", model)]

    return ImbPipeline(steps)

def get_proba(estimator, X):
    if hasattr(estimator, "predict_proba"):
        return estimator.predict_proba(X)[:, 1]
    if hasattr(estimator, "decision_function"):
        d = estimator.decision_function(X)
        return 1.0 / (1.0 + np.exp(-d))
    raise AttributeError("Estimator has neither predict_proba nor decision_function.")

def make_search_scorer(name: str):
    def _scorer(estimator, X, y):
        prob = get_proba(estimator, X)
        if name == "average_precision":
            return average_precision_score(y, prob)
        pred = (prob >= 0.5).astype(int)
        if name == "mcc_05":
            return matthews_corrcoef(y, pred) if len(np.unique(pred)) > 1 else -1.0
        if name == "balanced_accuracy_05":
            return balanced_accuracy_score(y, pred)
        if name == "roc_auc":
            return roc_auc_score(y, prob)
        raise ValueError(name)
    return _scorer

def make_group_cv(groups: pd.Series, requested_splits: int = 3):
    n_groups = groups.astype(str).nunique()
    n_splits = max(2, min(requested_splits, n_groups))
    return GroupKFold(n_splits=n_splits)

def tune_threshold(y_true: np.ndarray, prob: np.ndarray, metric: str = "mcc", specificity_floor: Optional[float] = None):
    rows = []
    best_thr = 0.5
    best_obj = -np.inf

    for thr in np.linspace(0.05, 0.95, 91):
        pred = (prob >= thr).astype(int)
        tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
        sens = tp / max(tp + fn, 1)
        spec = tn / max(tn + fp, 1)
        bacc = (sens + spec) / 2.0
        mcc = matthews_corrcoef(y_true, pred) if len(np.unique(pred)) > 1 else -1.0

        obj = bacc if metric == "balanced_accuracy" else mcc
        if specificity_floor is not None and spec < specificity_floor:
            obj = -np.inf

        rows.append({
            "threshold": thr,
            "sensitivity": sens,
            "specificity": spec,
            "balanced_accuracy": bacc,
            "mcc": mcc,
            "objective": obj,
        })

        if obj > best_obj:
            best_obj = obj
            best_thr = float(thr)

    return best_thr, pd.DataFrame(rows)

def get_oof_probabilities(estimator, X_train, y_train, groups):
    cv = make_group_cv(groups, N_INNER_SPLITS)
    prob_oof = np.zeros(len(X_train), dtype=float)

    for tr_idx, va_idx in cv.split(X_train, y_train, groups):
        est = clone(estimator)
        est.fit(X_train.iloc[tr_idx], y_train.iloc[tr_idx])
        prob_oof[va_idx] = get_proba(est, X_train.iloc[va_idx])

    return prob_oof

def evaluate_predictions(y_true, prob, thr):
    pred = (prob >= thr).astype(int)
    tn, fp, fn, tp = confusion_matrix(y_true, pred, labels=[0, 1]).ravel()
    sensitivity = tp / max(tp + fn, 1)
    specificity = tn / max(tn + fp, 1)
    balanced_accuracy = (sensitivity + specificity) / 2.0
    mcc = matthews_corrcoef(y_true, pred) if len(np.unique(pred)) > 1 else np.nan

    out = {
        "tn": int(tn), "fp": int(fp), "fn": int(fn), "tp": int(tp),
        "sensitivity": sensitivity,
        "specificity": specificity,
        "balanced_accuracy": balanced_accuracy,
        "mcc": mcc,
        "roc_auc": roc_auc_score(y_true, prob) if len(np.unique(y_true)) > 1 else np.nan,
        "pr_auc": average_precision_score(y_true, prob) if len(np.unique(y_true)) > 1 else np.nan,
    }
    return out, pred

def get_feature_names(fitted_pipeline, X_train, feature_cols):
    try:
        pre = fitted_pipeline.named_steps["preprocessor"]
        return pre.get_feature_names_out()
    except Exception:
        return np.array(feature_cols, dtype=object)
```

```python

# =========================
# 6. VISUALIZATION / ARTIFACT UTILITIES
# =========================
def save_confusion_plots(y_true, y_pred, outdir: Path):
    cm = confusion_matrix(y_true, y_pred, labels=[0, 1])
    cm_norm = cm.astype(float) / np.maximum(cm.sum(axis=1, keepdims=True), 1)

    pd.DataFrame(cm, index=["true_0", "true_1"], columns=["pred_0", "pred_1"]).to_csv(outdir / "confusion_matrix_counts.csv")
    pd.DataFrame(cm_norm, index=["true_0", "true_1"], columns=["pred_0", "pred_1"]).to_csv(outdir / "confusion_matrix_normalized.csv")

    for name, arr, fmt in [
        ("confusion_matrix.png", cm, "d"),
        ("confusion_matrix_normalized.png", cm_norm, ".2f"),
    ]:
        fig, ax = plt.subplots(figsize=(4.8, 4.0))
        im = ax.imshow(arr)
        ax.set_xticks([0, 1]); ax.set_yticks([0, 1])
        ax.set_xticklabels(["pred 0", "pred 1"]); ax.set_yticklabels(["true 0", "true 1"])
        ax.set_title(name.replace(".png", "").replace("_", " "))
        for i in range(2):
            for j in range(2):
                val = format(arr[i, j], fmt)
                ax.text(j, i, val, ha="center", va="center")
        fig.colorbar(im, ax=ax, shrink=0.8)
        fig.tight_layout()
        fig.savefig(outdir / name, bbox_inches="tight")
        plt.close(fig)

def save_roc_pr_plots(y_true, prob, outdir: Path):
    fpr, tpr, roc_thr = roc_curve(y_true, prob)
    prec, rec, pr_thr = precision_recall_curve(y_true, prob)

    pd.DataFrame({"fpr": fpr, "tpr": tpr, "threshold": np.append(roc_thr, np.nan)[:len(fpr)]}).to_csv(outdir / "roc_curve.csv", index=False)
    pd.DataFrame({"precision": prec, "recall": rec, "threshold": np.append(pr_thr, np.nan)[:len(prec)]}).to_csv(outdir / "pr_curve.csv", index=False)

    fig, ax = plt.subplots(figsize=(4.8, 4.0))
    ax.plot(fpr, tpr)
    ax.plot([0, 1], [0, 1], linestyle="--")
    ax.set_xlabel("FPR"); ax.set_ylabel("TPR"); ax.set_title("ROC curve")
    fig.tight_layout(); fig.savefig(outdir / "roc_curve.png", bbox_inches="tight"); plt.close(fig)

    fig, ax = plt.subplots(figsize=(4.8, 4.0))
    ax.plot(rec, prec)
    ax.set_xlabel("Recall"); ax.set_ylabel("Precision"); ax.set_title("PR curve")
    fig.tight_layout(); fig.savefig(outdir / "pr_curve.png", bbox_inches="tight"); plt.close(fig)

def save_threshold_sweep_plot(threshold_sweep_df: pd.DataFrame, outdir: Path):
    threshold_sweep_df.to_csv(outdir / "threshold_sweep.csv", index=False)
    fig, ax = plt.subplots(figsize=(5.6, 4.0))
    ax.plot(threshold_sweep_df["threshold"], threshold_sweep_df["mcc"], label="MCC")
    ax.plot(threshold_sweep_df["threshold"], threshold_sweep_df["balanced_accuracy"], label="BAcc")
    ax.plot(threshold_sweep_df["threshold"], threshold_sweep_df["specificity"], label="Specificity")
    ax.plot(threshold_sweep_df["threshold"], threshold_sweep_df["sensitivity"], label="Sensitivity")
    ax.set_xlabel("Threshold"); ax.set_ylabel("Score"); ax.set_title("Threshold sweep")
    ax.legend()
    fig.tight_layout(); fig.savefig(outdir / "threshold_sweep.png", bbox_inches="tight"); plt.close(fig)

def save_feature_importance(estimator, X_train, feature_names, outdir: Path):
    model = estimator.named_steps["model"]
    importances = None

    if hasattr(model, "feature_importances_"):
        importances = np.asarray(model.feature_importances_).ravel()
    elif hasattr(model, "coef_"):
        coef = np.asarray(model.coef_)
        importances = np.abs(coef.ravel())

    if importances is None:
        return

    n = min(len(feature_names), len(importances))
    feat_df = pd.DataFrame({
        "feature": np.asarray(feature_names[:n]).astype(str),
        "importance": importances[:n],
    }).sort_values("importance", ascending=False)

    feat_df.to_csv(outdir / "feature_importance.csv", index=False)
    top = feat_df.head(20).iloc[::-1]

    fig, ax = plt.subplots(figsize=(7.0, max(4.0, 0.28 * len(top))))
    ax.barh(top["feature"], top["importance"])
    ax.set_title("Top feature importance")
    fig.tight_layout()
    fig.savefig(outdir / "feature_importance_top.png", bbox_inches="tight")
    plt.close(fig)

def save_shap_artifacts(estimator, X_train, X_test, outdir: Path, max_samples: int = 300):
    if not SAVE_SHAP_DURING_TRAIN:
        return

    try:
        pre = estimator.named_steps["preprocessor"]
        model = estimator.named_steps["model"]

        Xtr_t = pre.transform(X_train)
        Xte_t = pre.transform(X_test)

        if hasattr(Xtr_t, "toarray"):
            Xtr_arr = Xtr_t.toarray()
        else:
            Xtr_arr = np.asarray(Xtr_t)

        if hasattr(Xte_t, "toarray"):
            Xte_arr = Xte_t.toarray()
        else:
            Xte_arr = np.asarray(Xte_t)

        if Xte_arr.shape[0] > max_samples:
            rng = np.random.default_rng(RANDOM_STATE)
            idx = rng.choice(np.arange(Xte_arr.shape[0]), size=max_samples, replace=False)
            Xte_arr = Xte_arr[idx]

        feature_names = list(pre.get_feature_names_out())

        if model.__class__.__name__.lower().startswith("xgb"):
            explainer = shap.TreeExplainer(model)
            shap_values = explainer.shap_values(Xte_arr)
        else:
            background = Xtr_arr[: min(200, len(Xtr_arr))]
            explainer = shap.LinearExplainer(model, background)
            shap_values = explainer.shap_values(Xte_arr)

        shap_arr = np.asarray(shap_values)
        if shap_arr.ndim == 3:
            shap_arr = shap_arr[..., 1]

        mean_abs = np.abs(shap_arr).mean(axis=0)
        shap_df = pd.DataFrame({"feature": feature_names, "mean_abs_shap": mean_abs}).sort_values("mean_abs_shap", ascending=False)
        shap_df.to_csv(outdir / "shap_mean_abs.csv", index=False)

        shap.summary_plot(shap_arr, Xte_arr, feature_names=feature_names, plot_type="bar", show=False, max_display=20)
        plt.tight_layout()
        plt.savefig(outdir / "shap_summary_bar.png", bbox_inches="tight")
        plt.close()

        shap.summary_plot(shap_arr, Xte_arr, feature_names=feature_names, show=False, max_display=20)
        plt.tight_layout()
        plt.savefig(outdir / "shap_beeswarm.png", bbox_inches="tight")
        plt.close()

    except Exception as e:
        (outdir / "shap_error.txt").write_text(str(e), encoding="utf-8")



def save_best_model_shap_from_disk(run_root: Path, max_samples: int = 200):
    """
    학습 중 SHAP을 끄고, best 모델들에 대해서만 후처리로 SHAP 생성.
    """
    if not SAVE_SHAP_FOR_BEST_ONLY:
        return
    best_path = run_root / "best_by_endpoint.csv"
    if not best_path.exists():
        print("[SHAP] best_by_endpoint.csv not found; skipping.")
        return

    try:
        import shap
    except Exception as e:
        print("[SHAP] import failed:", e)
        return

    best_df = pd.read_csv(best_path)
    for _, row in best_df.iterrows():
        exp_dir = Path(row["run_dir"])
        model_path = exp_dir / "best_estimator.joblib"
        meta_path = exp_dir / "artifact_manifest.json"
        pred_path = exp_dir / "test_predictions.csv"
        if (exp_dir / "shap_summary_bar.png").exists() and (exp_dir / "shap_beeswarm.png").exists():
            print(f"[SHAP] already exists -> {exp_dir.name}")
            continue
        if not model_path.exists() or not meta_path.exists():
            print(f"[SHAP] missing model/meta -> {exp_dir.name}")
            continue

        try:
            estimator = joblib.load(model_path)
            meta = json.loads(meta_path.read_text(encoding="utf-8"))
            endpoint = meta["endpoint"]
            feature_cols = meta["feature_cols"]

            train_df = loaded[endpoint]["train"]
            test_df = loaded[endpoint]["test"]

            X_train = train_df[feature_cols].copy()
            X_test = test_df[feature_cols].copy()

            pre = estimator.named_steps["preprocessor"]
            model = estimator.named_steps["model"]

            Xtr_t = pre.transform(X_train)
            Xte_t = pre.transform(X_test)

            if hasattr(Xtr_t, "toarray"):
                Xtr_arr = Xtr_t.toarray()
            else:
                Xtr_arr = np.asarray(Xtr_t)

            if hasattr(Xte_t, "toarray"):
                Xte_arr = Xte_t.toarray()
            else:
                Xte_arr = np.asarray(Xte_t)

            if Xte_arr.shape[0] > max_samples:
                rng = np.random.default_rng(RANDOM_STATE)
                idx = rng.choice(np.arange(Xte_arr.shape[0]), size=max_samples, replace=False)
                Xte_arr = Xte_arr[idx]

            feature_names = list(pre.get_feature_names_out())

            if model.__class__.__name__.lower().startswith("xgb"):
                explainer = shap.TreeExplainer(model)
                shap_values = explainer.shap_values(Xte_arr)
            else:
                background = Xtr_arr[: min(200, len(Xtr_arr))]
                explainer = shap.LinearExplainer(model, background)
                shap_values = explainer.shap_values(Xte_arr)

            shap_arr = np.asarray(shap_values)
            if shap_arr.ndim == 3:
                shap_arr = shap_arr[..., 1]

            mean_abs = np.abs(shap_arr).mean(axis=0)
            shap_df = pd.DataFrame({"feature": feature_names, "mean_abs_shap": mean_abs}).sort_values("mean_abs_shap", ascending=False)
            shap_df.to_csv(exp_dir / "shap_mean_abs.csv", index=False)

            shap.summary_plot(shap_arr, Xte_arr, feature_names=feature_names, plot_type="bar", show=False, max_display=20)
            plt.tight_layout()
            plt.savefig(exp_dir / "shap_summary_bar.png", bbox_inches="tight")
            plt.close()

            shap.summary_plot(shap_arr, Xte_arr, feature_names=feature_names, show=False, max_display=20)
            plt.tight_layout()
            plt.savefig(exp_dir / "shap_beeswarm.png", bbox_inches="tight")
            plt.close()

            print(f"[SHAP] saved -> {exp_dir}")
            safe_gc()

        except Exception as e:
            (exp_dir / "shap_error.txt").write_text(str(e), encoding="utf-8")
            print(f"[SHAP] failed -> {exp_dir.name}: {e}")
            safe_gc()

def build_summary_dashboard(summary_df: pd.DataFrame, outdir: Path):
    if summary_df.empty:
        return

    summary_df.to_csv(outdir / "summary_metrics.csv", index=False)

    ranked = summary_df.sort_values(["endpoint", "mcc", "balanced_accuracy", "pr_auc"], ascending=[True, False, False, False]).copy()
    ranked.to_csv(outdir / "summary_metrics_ranked.csv", index=False)

    best = ranked.groupby("endpoint", as_index=False).head(1).copy()
    best.to_csv(outdir / "best_by_endpoint.csv", index=False)

    # overview dashboard
    fig, axes = plt.subplots(2, 2, figsize=(14, 10))
    plot_metrics = [("mcc", "MCC"), ("balanced_accuracy", "Balanced accuracy"), ("roc_auc", "ROC-AUC"), ("pr_auc", "PR-AUC")]

    for ax, (metric, title) in zip(axes.ravel(), plot_metrics):
        pivot = ranked.pivot_table(index="endpoint", columns="label", values=metric, aggfunc="first")
        pivot.plot(kind="bar", ax=ax)
        ax.set_title(title)
        ax.legend(loc="best", fontsize=8)
        ax.tick_params(axis="x", rotation=0)

    fig.suptitle("Summary overview dashboard", fontsize=16, fontweight="bold")
    fig.tight_layout()
    fig.savefig(outdir / "summary_overview_dashboard.png", bbox_inches="tight")
    plt.close(fig)

    # best artifact gallery
    fig, axes = plt.subplots(len(best), 2, figsize=(10, 4 * len(best)))
    if len(best) == 1:
        axes = np.array([axes])
    for i, (_, row) in enumerate(best.iterrows()):
        exp_dir = Path(row["run_dir"])
        left = axes[i, 0]
        right = axes[i, 1]
        left.set_title(f"{row['endpoint']} | {row['label']}\nConfusion matrix")
        right.set_title(f"{row['endpoint']} | {row['label']}\nSHAP / importance")
        left.axis("off"); right.axis("off")

        cm_path = exp_dir / "confusion_matrix.png"
        shap_path = exp_dir / "shap_summary_bar.png"
        fi_path = exp_dir / "feature_importance_top.png"

        if cm_path.exists():
            left.imshow(plt.imread(cm_path))
        if shap_path.exists():
            right.imshow(plt.imread(shap_path))
        elif fi_path.exists():
            right.imshow(plt.imread(fi_path))

    fig.tight_layout()
    fig.savefig(outdir / "best_artifact_gallery.png", bbox_inches="tight")
    plt.close(fig)
```

```python

# =========================
# 7. RUN SINGLE EXPERIMENT
# =========================
def run_single_experiment(endpoint: str, model_name: str, strategy: str, run_root: Path):
    train_df = loaded[endpoint]["train"].copy()
    test_df = loaded[endpoint]["test"].copy()
    feature_cols = list(loaded[endpoint]["feature_cols"])

    X_train = train_df[feature_cols].copy()
    X_test = test_df[feature_cols].copy()
    y_train = train_df[TARGET_COL].astype(int)
    y_test = test_df[TARGET_COL].astype(int)
    groups = train_df[GROUP_COL].astype(str)

    label = f"{endpoint}__{model_name}__{strategy}"
    exp_dir = run_root / label
    exp_dir.mkdir(parents=True, exist_ok=True)

    # 완료된 실험은 skip
    if SKIP_COMPLETED and (exp_dir / "metrics.json").exists():
        try:
            row = json.loads((exp_dir / "metrics.json").read_text(encoding="utf-8"))
            row["skipped_existing"] = True
            print(f"[SKIP] {label}")
            return row
        except Exception:
            pass

    try:
        pipe = build_pipeline(X_train, feature_cols, endpoint, model_name, strategy)

        if GRIDSEARCH_ENABLED:
            grid = GridSearchCV(
                estimator=pipe,
                param_grid=get_param_grid(model_name, endpoint),
                cv=make_group_cv(groups, N_INNER_SPLITS),
                scoring=make_search_scorer(ENDPOINT_MODEL_CFG[endpoint]["grid_scorer"]),
                refit=True,
                n_jobs=1,
                verbose=0,
                error_score="raise",
            )
            grid.fit(X_train, y_train, groups=groups)
            best_estimator = grid.best_estimator_
            best_params = grid.best_params_
            best_cv_score = float(grid.best_score_)
        else:
            best_estimator = clone(pipe).fit(X_train, y_train)
            best_params = {}
            best_cv_score = np.nan

        thr_metric = ENDPOINT_MODEL_CFG[endpoint]["threshold_metric"]
        spec_floor = ENDPOINT_MODEL_CFG[endpoint]["threshold_specificity_floor"]
        prob_oof = get_oof_probabilities(best_estimator, X_train, y_train, groups)
        threshold, sweep_df = tune_threshold(y_train.to_numpy(), prob_oof, metric=thr_metric, specificity_floor=spec_floor)

        best_estimator.fit(X_train, y_train)
        test_prob = get_proba(best_estimator, X_test)
        metrics, test_pred = evaluate_predictions(y_test.to_numpy(), test_prob, threshold)

        feature_names = get_feature_names(best_estimator, X_train, feature_cols)
        metrics_row = {
            "endpoint": endpoint,
            "model_name": model_name,
            "resampling_strategy": strategy,
            "label": label,
            "n_features": len(feature_cols),
            "threshold": threshold,
            "threshold_metric": thr_metric,
            "specificity_floor": spec_floor,
            "grid_scorer": ENDPOINT_MODEL_CFG[endpoint]["grid_scorer"],
            "gridsearch_enabled": GRIDSEARCH_ENABLED,
            "best_cv_score": best_cv_score,
            "train_n": len(train_df),
            "test_n": len(test_df),
            "train_pos_rate": float(y_train.mean()),
            "test_pos_rate": float(y_test.mean()),
            "run_dir": str(exp_dir),
            **metrics,
        }

        # save predictions
        pred_df = test_df[[c for c in test_df.columns if c in ["No", "SMILES", TARGET_COL, GROUP_COL, "murcko_scaffold"]]].copy()
        pred_df["prob_pos"] = test_prob
        pred_df["pred"] = test_pred
        pred_df["threshold"] = threshold
        pred_df.to_csv(exp_dir / "test_predictions.csv", index=False)
        pred_df[pred_df[TARGET_COL] != pred_df["pred"]].to_csv(exp_dir / "test_misclassifications.csv", index=False)

        # save estimator + artifacts
        joblib.dump(best_estimator, exp_dir / "best_estimator.joblib", compress=3)
        (exp_dir / "metrics.json").write_text(json.dumps(metrics_row, indent=2), encoding="utf-8")
        (exp_dir / "best_params.json").write_text(json.dumps(best_params, indent=2, default=str), encoding="utf-8")
        save_confusion_plots(y_test.to_numpy(), test_pred, exp_dir)
        save_roc_pr_plots(y_test.to_numpy(), test_prob, exp_dir)
        save_threshold_sweep_plot(sweep_df, exp_dir)
        save_feature_importance(best_estimator, X_train, feature_names, exp_dir)
        if SAVE_SHAP_DURING_TRAIN:
            save_shap_artifacts(best_estimator, X_train, X_test, exp_dir)

        # manifest
        manifest = {
            "endpoint": endpoint,
            "model_name": model_name,
            "resampling_strategy": strategy,
            "feature_cols": feature_cols,
            "n_features": len(feature_cols),
            "best_params": best_params,
            "threshold": threshold,
        }
        (exp_dir / "artifact_manifest.json").write_text(json.dumps(manifest, indent=2), encoding="utf-8")

        print(f"[DONE] {label} | MCC={metrics_row['mcc']:.3f} | BAcc={metrics_row['balanced_accuracy']:.3f} | ROC-AUC={metrics_row['roc_auc']:.3f}")
        print("       saved ->", exp_dir)
        return metrics_row

    except Exception as e:
        err = {
            "endpoint": endpoint,
            "model_name": model_name,
            "resampling_strategy": strategy,
            "label": label,
            "run_dir": str(exp_dir),
            "error": str(e),
            "traceback": traceback.format_exc(),
        }
        (exp_dir / "error.txt").write_text(err["traceback"], encoding="utf-8")
        print(f"[FAIL] {label}: {e}")
        return err
    finally:
        safe_gc()
```

```python

# =========================
# 8. RUN BATCHED EXPERIMENTS (RESUMABLE)
# =========================
timestamp = pd.Timestamp.now().strftime("%Y%m%d_%H%M%S")
latest_pointer = RUNS_DIR / "LATEST_RESAMPLE_RUN_PATH.txt"

if latest_pointer.exists():
    last_run = Path(latest_pointer.read_text(encoding="utf-8").strip())
    if last_run.exists():
        run_root = last_run
    else:
        run_root = RUNS_DIR / f"allinone_resample_run_{timestamp}"
else:
    run_root = RUNS_DIR / f"allinone_resample_run_{timestamp}"

run_root.mkdir(parents=True, exist_ok=True)
latest_pointer.write_text(str(run_root), encoding="utf-8")

summary_path = run_root / "summary_metrics.csv"
failed_path = run_root / "failed_experiments.csv"

existing_rows = []
if summary_path.exists():
    try:
        existing_rows = pd.read_csv(summary_path).to_dict("records")
    except Exception:
        existing_rows = []

completed = set()
for row in existing_rows:
    completed.add((row.get("endpoint"), row.get("model_name"), row.get("resampling_strategy")))

plan = []
for endpoint in ENDPOINTS:
    for model_name in MODELS_TO_RUN:
        for strategy in RESAMPLING_STRATEGIES:
            triplet = (endpoint, model_name, strategy)
            if SKIP_COMPLETED and triplet in completed:
                continue
            plan.append(triplet)

print(f"Run root: {run_root}")
print(f"Pending experiments: {len(plan)}")
if plan:
    print("Next batch:", plan[:EXPERIMENTS_PER_BATCH])

batch = plan[:EXPERIMENTS_PER_BATCH]
all_rows = list(existing_rows)

for endpoint, model_name, strategy in batch:
    row = run_single_experiment(endpoint, model_name, strategy, run_root)
    all_rows = [r for r in all_rows if not (
        r.get("endpoint") == endpoint and r.get("model_name") == model_name and r.get("resampling_strategy") == strategy
    )]
    all_rows.append(row)

    # partial save after each experiment
    df_now = pd.DataFrame(all_rows)
    df_now.to_csv(summary_path, index=False)

    fails = df_now[df_now["error"].notna()] if "error" in df_now.columns else pd.DataFrame()
    if not fails.empty:
        fails.to_csv(failed_path, index=False)

    safe_gc()

summary_df = pd.read_csv(summary_path) if summary_path.exists() else pd.DataFrame()
display(summary_df)

if AUTO_SUMMARY_AFTER_BATCH and not summary_df.empty:
    build_summary_dashboard(summary_df, run_root)

(run_root / "run_info.json").write_text(json.dumps({
    "base_dir": str(BASE_DIR),
    "data_dir": str(DATA_DIR),
    "feature_blocks": FEATURE_BLOCKS,
    "fp_top_k": FP_TOP_K,
    "qm_file": str(detect_qm_file()) if detect_qm_file() else None,
    "endpoints": ENDPOINTS,
    "models": MODELS_TO_RUN,
    "resampling_strategies": RESAMPLING_STRATEGIES,
    "gridsearch_enabled": GRIDSEARCH_ENABLED,
    "n_inner_splits": N_INNER_SPLITS,
    "quick_mode": QUICK_MODE,
    "save_shap_during_train": SAVE_SHAP_DURING_TRAIN,
    "save_shap_for_best_only": SAVE_SHAP_FOR_BEST_ONLY,
    "experiments_per_batch": EXPERIMENTS_PER_BATCH,
}, indent=2), encoding="utf-8")

pending_after = max(0, len(plan) - len(batch))
print("\n[PARTIAL COMPLETE] run root:", run_root)
print("Processed in this batch:", len(batch))
print("Remaining pending:", pending_after)
if pending_after > 0:
    print("이 셀을 다시 한 번 실행하면 다음 배치를 이어서 수행합니다.")

if SAVE_SHAP_FOR_BEST_ONLY and AUTO_SUMMARY_AFTER_BATCH and pending_after == 0:
    print("\nAll experiments finished. Generating SHAP for best models only...")
    save_best_model_shap_from_disk(run_root, max_samples=BEST_SHAP_MAX_SAMPLES)
    build_summary_dashboard(pd.read_csv(summary_path), run_root)
```

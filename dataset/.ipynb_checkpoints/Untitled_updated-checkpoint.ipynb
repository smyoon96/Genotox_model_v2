```python
# =========================================================
ALL_FG_SMARTS = {}
ALL_FG_SMARTS.update(FUNCTIONAL_GROUP_SMARTS)
ALL_FG_SMARTS.update(BENIGNI_BOSSA_ALERTS)

FG_PATTERNS = {}
for name, smarts in ALL_FG_SMARTS.items():
    patt = Chem.MolFromSmarts(smarts)
    if patt is None:
        print(f"[WARNING] Invalid SMARTS for {name}: {smarts}")
    FG_PATTERNS[name] = patt
```

```python
# =========================================================
# 4. 분자 변환
```

```python
# =========================================================
def load_dataset(path):
    ext = os.path.splitext(path)[1].lower()
    if ext == ".csv":
        return pd.read_csv(path)
    elif ext in [".xlsx", ".xls"]:
        return pd.read_excel(path)
    raise ValueError(f"Unsupported file type: {ext}")


def mol_from_smiles(smiles):
    if pd.isna(smiles):
        return None
    s = str(smiles).strip()
    if not s:
        return None
    try:
        return Chem.MolFromSmiles(s)
    except Exception:
        return None


def canonical_only_smiles(smiles):
    mol = mol_from_smiles(smiles)
    if mol is None:
        return None
    try:
        return Chem.MolToSmiles(mol, canonical=True)
    except Exception:
        return None


def standardized_parent_smiles(smiles):
    mol = mol_from_smiles(smiles)
    if mol is None:
        return None
    try:
        clean = rdMolStandardize.Cleanup(mol)
        parent = rdMolStandardize.FragmentParent(clean)  # salt 제거
        parent = rdMolStandardize.Uncharger().uncharge(parent)
        return Chem.MolToSmiles(parent, canonical=True)
    except Exception:
        return None
```

```python
# =========================================================
# 5. 금속 / multi-component / 이온성 체크
```

```python
# =========================================================
def get_element_counts(mol):
    c = Counter()
    if mol is None:
        return c
    for atom in mol.GetAtoms():
        c[atom.GetSymbol()] += 1
    return c


def get_fragment_count_from_smiles(smiles):
    if smiles is None or pd.isna(smiles):
        return np.nan
    return str(smiles).count(".") + 1


def contains_metal(mol):
    if mol is None:
        return 0
    return int(any(atom.GetSymbol() in METAL_ELEMENTS for atom in mol.GetAtoms()))


def contains_heavy_metal(mol):
    if mol is None:
        return 0
    return int(any(atom.GetSymbol() in HEAVY_METAL_ELEMENTS for atom in mol.GetAtoms()))


def count_metal_atoms(mol):
    if mol is None:
        return np.nan
    return sum(atom.GetSymbol() in METAL_ELEMENTS for atom in mol.GetAtoms())


def count_heavy_metal_atoms(mol):
    if mol is None:
        return np.nan
    return sum(atom.GetSymbol() in HEAVY_METAL_ELEMENTS for atom in mol.GetAtoms())


def metal_element_string(mol):
    if mol is None:
        return None
    elems = sorted({atom.GetSymbol() for atom in mol.GetAtoms() if atom.GetSymbol() in METAL_ELEMENTS})
    return ",".join(elems) if elems else None


def has_formal_charge(mol):
    if mol is None:
        return np.nan
    return int(any(atom.GetFormalCharge() != 0 for atom in mol.GetAtoms()))


def total_formal_charge(mol):
    if mol is None:
        return np.nan
    return int(sum(atom.GetFormalCharge() for atom in mol.GetAtoms()))


def has_counterion_like_pattern(mol):
    if mol is None:
        return 0
    elems = {atom.GetSymbol() for atom in mol.GetAtoms()}
    return int(len(elems.intersection(COMMON_COUNTERION_ELEMENTS)) > 0 and "." in Chem.MolToSmiles(mol))
```

```python
# =========================================================
# 6. Descriptor 계산
```

```python
# =========================================================
def safe_calc(fn, mol):
    try:
        return fn(mol)
    except Exception:
        return np.nan


def calc_descriptors(mol):
    if mol is None:
        return {
            "mw": np.nan,
            "logp": np.nan,
            "tpsa": np.nan,
            "hbd": np.nan,
            "hba": np.nan,
            "rot_bonds": np.nan,
            "ring_count": np.nan,
            "aromatic_ring_count": np.nan,
            "heteroatom_count": np.nan,
            "fraction_csp3": np.nan,
            "heavy_atom_count": np.nan,
            "formal_charge_total": np.nan,
            "has_formal_charge": np.nan,
        }

    return {
        "mw": safe_calc(Descriptors.MolWt, mol),
        "logp": safe_calc(Crippen.MolLogP, mol),
        "tpsa": safe_calc(rdMolDescriptors.CalcTPSA, mol),
        "hbd": safe_calc(Lipinski.NumHDonors, mol),
        "hba": safe_calc(Lipinski.NumHAcceptors, mol),
        "rot_bonds": safe_calc(Lipinski.NumRotatableBonds, mol),
        "ring_count": safe_calc(rdMolDescriptors.CalcNumRings, mol),
        "aromatic_ring_count": safe_calc(rdMolDescriptors.CalcNumAromaticRings, mol),
        "heteroatom_count": safe_calc(Lipinski.NumHeteroatoms, mol),
        "fraction_csp3": safe_calc(rdMolDescriptors.CalcFractionCSP3, mol),
        "heavy_atom_count": safe_calc(Lipinski.HeavyAtomCount, mol),
        "formal_charge_total": total_formal_charge(mol),
        "has_formal_charge": has_formal_charge(mol),
    }
```

```python
# =========================================================
# 7. 작용기 카운트 (일반 FG + Benigni-Bossa SA 통합)
```

```python
# =========================================================
def functional_group_counts(mol):
    out = {}
    if mol is None:
        for k in FG_PATTERNS:
            out[f"fg_{k}_count"] = np.nan
            out[f"fg_{k}_present"] = np.nan
        return out

    for name, patt in FG_PATTERNS.items():
        if patt is None:
            out[f"fg_{name}_count"] = np.nan
            out[f"fg_{name}_present"] = np.nan
            continue
        try:
            matches = mol.GetSubstructMatches(patt)
            n = len(matches)
        except Exception:
            n = np.nan

        out[f"fg_{name}_count"] = n
        out[f"fg_{name}_present"] = int(n > 0) if pd.notna(n) else np.nan

    return out
```

```python
# =========================================================
# 7-1. Benigni-Bossa SA 요약 (총 SA 수, 유전독성/비유전독성 분리)
```

```python
# =========================================================
def benigni_bossa_summary(mol):
    """Benigni-Bossa SA 관련 집약 feature를 반환"""
    result = {
        "bb_any_alert": 0,
        "bb_n_alerts": 0,
        "bb_any_genotox_alert": 0,
        "bb_n_genotox_alerts": 0,
        "bb_any_nongenotox_alert": 0,
        "bb_n_nongenotox_alerts": 0,
    }
    if mol is None:
        return {k: np.nan for k in result}

    # Genotoxic alerts: SA1 ~ SA32
    genotox_prefixes = [
        "BB_SA1_", "BB_SA2", "BB_SA3_", "BB_SA4_", "BB_SA5_",
        "BB_SA6", "BB_SA7_", "BB_SA8_", "BB_SA9_", "BB_SA10_",
        "BB_SA11", "BB_SA12", "BB_SA13_", "BB_SA14", "BB_SA15_",
        "BB_SA16", "BB_SA17", "BB_SA18", "BB_SA19", "BB_SA20_",
        "BB_SA21_", "BB_SA22_", "BB_SA23_", "BB_SA24", "BB_SA25_",
        "BB_SA26", "BB_SA27", "BB_SA28_", "BB_SA29", "BB_SA30_",
        "BB_SA31_", "BB_SA32_",
    ]
    # Nongenotoxic alerts: SA33 ~ SA55
    nongenotox_prefixes = [
        "BB_SA33_", "BB_SA34_", "BB_SA35", "BB_SA36_", "BB_SA37_",
        "BB_SA38_", "BB_SA39_", "BB_SA40_", "BB_SA41", "BB_SA42_",
        "BB_SA43_", "BB_SA44", "BB_SA45", "BB_SA46_", "BB_SA47_",
        "BB_SA48_", "BB_SA49_", "BB_SA50_", "BB_SA51_", "BB_SA52",
        "BB_SA54_", "BB_SA55_",
    ]

    n_genotox = 0
    n_nongenotox = 0

    for name, patt in FG_PATTERNS.items():
        if not name.startswith("BB_"):
            continue
        if patt is None:
            continue
        try:
            if mol.HasSubstructMatch(patt):
                is_genotox = any(name.startswith(p) for p in genotox_prefixes)
                is_nongenotox = any(name.startswith(p) for p in nongenotox_prefixes)
                if is_genotox:
                    n_genotox += 1
                if is_nongenotox:
                    n_nongenotox += 1
        except Exception:
            pass

    result["bb_n_genotox_alerts"] = n_genotox
    result["bb_any_genotox_alert"] = int(n_genotox > 0)
    result["bb_n_nongenotox_alerts"] = n_nongenotox
    result["bb_any_nongenotox_alert"] = int(n_nongenotox > 0)
    result["bb_n_alerts"] = n_genotox + n_nongenotox
    result["bb_any_alert"] = int((n_genotox + n_nongenotox) > 0)

    return result
```

```python
# =========================================================
# 8. Scaffold / 기본 feature
```

```python
# =========================================================
def murcko_scaffold_smiles(mol):
    if mol is None:
        return None
    try:
        scaf = MurckoScaffold.GetScaffoldForMol(mol)
        if scaf is None or scaf.GetNumAtoms() == 0:
            return None
        return Chem.MolToSmiles(scaf, canonical=True)
    except Exception:
        return None


def basic_structure_flags(smiles, mol):
    return {
        "fragment_count": get_fragment_count_from_smiles(smiles),
        "is_multicomponent": int(get_fragment_count_from_smiles(smiles) > 1) if smiles is not None else np.nan,
        "contains_metal": contains_metal(mol),
        "contains_heavy_metal": contains_heavy_metal(mol),
        "metal_atom_count": count_metal_atoms(mol),
        "heavy_metal_atom_count": count_heavy_metal_atoms(mol),
        "metal_elements": metal_element_string(mol),
        "has_counterion_like": has_counterion_like_pattern(mol),
        "murcko_scaffold": murcko_scaffold_smiles(mol),
    }
```

```python
# =========================================================
# 9. 하나의 representation 분석 테이블 생성
```

```python
# =========================================================
def build_representation_analysis_df(df, smiles_col, rep_name):
    work = df.copy()
    work["analysis_smiles"] = work[smiles_col].astype(str)
    work["mol"] = work["analysis_smiles"].apply(mol_from_smiles)

    rows = []
    for _, row in work.iterrows():
        mol = row["mol"]
        smi = row["analysis_smiles"]

        base = {
            "No": row["No"] if "No" in row else np.nan,
            "label": row[LABEL_COL],
            "representation": rep_name,
            "analysis_smiles": smi,
            "valid": int(mol is not None),
        }

        base.update(basic_structure_flags(smi, mol))
        base.update(calc_descriptors(mol))
        base.update(functional_group_counts(mol))
        base.update(benigni_bossa_summary(mol))

        rows.append(base)

    out = pd.DataFrame(rows)
    return out
```

```python
# =========================================================
# 10. 통계 비교
```

```python
# =========================================================
def summarize_binary_features(df, group_col="label"):
    present_cols = [c for c in df.columns if c.endswith("_present")]
    extra_binary = [
        "is_multicomponent", "contains_metal", "contains_heavy_metal",
        "has_counterion_like", "has_formal_charge",
        "bb_any_alert", "bb_any_genotox_alert", "bb_any_nongenotox_alert",
    ]
    binary_cols = [c for c in present_cols + extra_binary if c in df.columns]

    records = []
    sub = df[df["valid"] == 1].copy()

    for col in binary_cols:
        tmp = sub[[group_col, col]].dropna()
        if tmp.empty or tmp[group_col].nunique() < 2:
            continue

        pos = tmp[tmp[group_col] == 1][col]
        neg = tmp[tmp[group_col] == 0][col]

        pos_rate = pos.mean() if len(pos) > 0 else np.nan
        neg_rate = neg.mean() if len(neg) > 0 else np.nan

        # 2x2 contingency
        table = pd.crosstab(tmp[group_col], tmp[col])
        pval = np.nan
        test_used = None

        try:
            if table.shape == (2, 2):
                if (table.values < 5).any():
                    _, pval = fisher_exact(table.values)
                    test_used = "fisher"
                else:
                    _, pval, _, _ = chi2_contingency(table.values)
                    test_used = "chi2"
        except Exception:
            pass

        records.append({
            "feature": col,
            "positive_rate": pos_rate,
            "negative_rate": neg_rate,
            "rate_diff": pos_rate - neg_rate if pd.notna(pos_rate) and pd.notna(neg_rate) else np.nan,
            "test": test_used,
            "p_value": pval,
            "n_positive": len(pos),
            "n_negative": len(neg),
        })

    return pd.DataFrame(records).sort_values(["p_value", "rate_diff"], ascending=[True, False])


def summarize_numeric_features(df, group_col="label"):
    num_cols = [
        "fragment_count", "metal_atom_count", "heavy_metal_atom_count",
        "mw", "logp", "tpsa", "hbd", "hba", "rot_bonds", "ring_count",
        "aromatic_ring_count", "heteroatom_count", "fraction_csp3",
        "heavy_atom_count", "formal_charge_total",
        "bb_n_alerts", "bb_n_genotox_alerts", "bb_n_nongenotox_alerts",
    ]
    num_cols = [c for c in num_cols if c in df.columns]

    records = []
    sub = df[df["valid"] == 1].copy()

    for col in num_cols:
        tmp = sub[[group_col, col]].dropna()
        if tmp.empty or tmp[group_col].nunique() < 2:
            continue

        pos = tmp[tmp[group_col] == 1][col]
        neg = tmp[tmp[group_col] == 0][col]

        pval = np.nan
        try:
            if len(pos) > 0 and len(neg) > 0:
                _, pval = mannwhitneyu(pos, neg, alternative="two-sided")
        except Exception:
            pass

        records.append({
            "feature": col,
            "positive_median": pos.median() if len(pos) else np.nan,
            "negative_median": neg.median() if len(neg) else np.nan,
            "positive_mean": pos.mean() if len(pos) else np.nan,
            "negative_mean": neg.mean() if len(neg) else np.nan,
            "median_diff": (pos.median() - neg.median()) if len(pos) and len(neg) else np.nan,
            "test": "mannwhitneyu",
            "p_value": pval,
            "n_positive": len(pos),
            "n_negative": len(neg),
        })

    return pd.DataFrame(records).sort_values(["p_value", "median_diff"], ascending=[True, False])


def descriptor_label_correlations(df, group_col="label"):
    cols = [
        "mw", "logp", "tpsa", "hbd", "hba", "rot_bonds", "ring_count",
        "aromatic_ring_count", "heteroatom_count", "fraction_csp3",
        "heavy_atom_count", "metal_atom_count", "heavy_metal_atom_count",
        "fragment_count", "formal_charge_total",
        "bb_n_alerts", "bb_n_genotox_alerts", "bb_n_nongenotox_alerts",
    ]
    cols = [c for c in cols if c in df.columns]

    sub = df[df["valid"] == 1].copy()
    records = []

    for col in cols:
        tmp = sub[[group_col, col]].dropna()
        if tmp.empty or tmp[group_col].nunique() < 2:
            continue

        try:
            rho, pval = spearmanr(tmp[group_col], tmp[col])
        except Exception:
            rho, pval = np.nan, np.nan

        records.append({
            "feature": col,
            "spearman_rho_with_label": rho,
            "p_value": pval,
            "n": len(tmp),
        })

    return pd.DataFrame(records).sort_values("p_value", ascending=True)
```

```python
# =========================================================
# 10-1. 양성/음성 그룹 내 작용기 ↔ 물리화학적 특성 상관분석
```

```python
# =========================================================
def intra_group_correlation(df, group_col="label"):
    """
    양성(label=1)과 음성(label=0) 그룹 각각에 대해:
      - 작용기 (fg_*_present, fg_*_count, bb_* ) ↔ 물리화학적 특성
      의 Spearman 상관계수를 산출
    결과: 그룹별 feature-pair 상관 테이블
    """
    descriptor_cols = [
        "mw", "logp", "tpsa", "hbd", "hba", "rot_bonds", "ring_count",
        "aromatic_ring_count", "heteroatom_count", "fraction_csp3",
        "heavy_atom_count",
    ]
    descriptor_cols = [c for c in descriptor_cols if c in df.columns]

    # 작용기 count 컬럼 (0이 아닌 값이 존재하는 것만)
    fg_count_cols = [c for c in df.columns if c.endswith("_count") and c.startswith("fg_")]
    # Benigni-Bossa 집약 컬럼
    bb_cols = [c for c in df.columns if c.startswith("bb_") and c in df.columns]

    feature_cols = fg_count_cols + bb_cols

    sub = df[df["valid"] == 1].copy()
    all_records = []

    for label_val in [1, 0]:
        group_name = "positive" if label_val == 1 else "negative"
        grp = sub[sub[group_col] == label_val]

        if len(grp) < 10:
            continue

        for fc in feature_cols:
            # 최소 5% 이상 non-zero가 있어야 상관분석 의미 있음
            if fc not in grp.columns:
                continue
            fc_vals = grp[fc].dropna()
            if fc_vals.nunique() < 2:
                continue
            nonzero_rate = (fc_vals != 0).mean()
            if nonzero_rate < 0.01:
                continue

            for dc in descriptor_cols:
                tmp = grp[[fc, dc]].dropna()
                if len(tmp) < 10:
                    continue
                try:
                    rho, pval = spearmanr(tmp[fc], tmp[dc])
                except Exception:
                    rho, pval = np.nan, np.nan

                all_records.append({
                    "group": group_name,
                    "fg_feature": fc,
                    "descriptor": dc,
                    "spearman_rho": rho,
                    "p_value": pval,
                    "n": len(tmp),
                })

    result = pd.DataFrame(all_records)
    if not result.empty:
        result = result.sort_values(["group", "p_value"], ascending=[True, True])
    return result


def intra_group_descriptor_correlation(df, group_col="label"):
    """
    양성/음성 그룹 각각 내에서 물리화학적 특성 간 Spearman 상관행렬
    """
    descriptor_cols = [
        "mw", "logp", "tpsa", "hbd", "hba", "rot_bonds", "ring_count",
        "aromatic_ring_count", "heteroatom_count", "fraction_csp3",
        "heavy_atom_count", "bb_n_alerts", "bb_n_genotox_alerts", "bb_n_nongenotox_alerts",
    ]
    descriptor_cols = [c for c in descriptor_cols if c in df.columns]

    sub = df[df["valid"] == 1].copy()
    all_records = []

    for label_val in [1, 0]:
        group_name = "positive" if label_val == 1 else "negative"
        grp = sub[sub[group_col] == label_val]

        if len(grp) < 10:
            continue

        for i, c1 in enumerate(descriptor_cols):
            for c2 in descriptor_cols[i+1:]:
                tmp = grp[[c1, c2]].dropna()
                if len(tmp) < 10:
                    continue
                try:
                    rho, pval = spearmanr(tmp[c1], tmp[c2])
                except Exception:
                    rho, pval = np.nan, np.nan

                all_records.append({
                    "group": group_name,
                    "feature_1": c1,
                    "feature_2": c2,
                    "spearman_rho": rho,
                    "p_value": pval,
                    "n": len(tmp),
                })

    result = pd.DataFrame(all_records)
    if not result.empty:
        result = result.sort_values(["group", "p_value"], ascending=[True, True])
    return result
```

```python
# =========================================================
# 10-2. 기존 통계 함수들 (metal, shift)
```

```python
# =========================================================
def metal_summary(df):
    sub = df[df["valid"] == 1].copy()

    total = len(sub)
    if total == 0:
        return pd.DataFrame([{
            "n_valid": 0,
            "contains_metal_rate": np.nan,
            "contains_heavy_metal_rate": np.nan,
            "multicomponent_rate": np.nan,
        }])

    metal_rows = sub[sub["contains_metal"] == 1].copy()

    elem_counter = Counter()
    for x in metal_rows["metal_elements"].dropna():
        for e in str(x).split(","):
            if e:
                elem_counter[e] += 1

    top_elems = "; ".join([f"{k}:{v}" for k, v in elem_counter.most_common(20)])

    return pd.DataFrame([{
        "n_valid": total,
        "n_contains_metal": int(sub["contains_metal"].sum()),
        "n_contains_heavy_metal": int(sub["contains_heavy_metal"].sum()),
        "contains_metal_rate": sub["contains_metal"].mean(),
        "contains_heavy_metal_rate": sub["contains_heavy_metal"].mean(),
        "multicomponent_rate": sub["is_multicomponent"].mean(),
        "charged_rate": sub["has_formal_charge"].mean(),
        "top_metal_elements": top_elems,
    }])


def compare_preprocessing_shift(df_canon, df_std, join_cols=("No", "label")):
    merged = df_canon.merge(
        df_std,
        on=list(join_cols),
        suffixes=("_canon", "_std"),
        how="inner"
    )

    cols = [
        "mw", "logp", "tpsa", "hbd", "hba", "rot_bonds", "ring_count",
        "aromatic_ring_count", "heteroatom_count", "fraction_csp3",
        "heavy_atom_count", "fragment_count", "metal_atom_count",
        "heavy_metal_atom_count", "formal_charge_total"
    ]

    records = []
    for col in cols:
        c1 = f"{col}_canon"
        c2 = f"{col}_std"
        if c1 not in merged.columns or c2 not in merged.columns:
            continue

        tmp = merged[[c1, c2]].dropna()
        if tmp.empty:
            continue

        diff = tmp[c2] - tmp[c1]
        records.append({
            "feature": col,
            "n": len(tmp),
            "canon_mean": tmp[c1].mean(),
            "std_mean": tmp[c2].mean(),
            "mean_shift_std_minus_canon": diff.mean(),
            "median_shift_std_minus_canon": diff.median(),
        })

    return pd.DataFrame(records).sort_values("mean_shift_std_minus_canon", ascending=False)
```

```python
# =========================================================
# 11. 메인 실행
```

```python
# =========================================================
def prepare_three_views(df):
    out = df.copy()
    out["raw_smiles"] = out[SMILES_COL].astype(str)
    out["canonical_smiles"] = out["raw_smiles"].apply(canonical_only_smiles)
    out["standardized_smiles"] = out["raw_smiles"].apply(standardized_parent_smiles)
    return out


def run_analysis_for_file(path):
    name = os.path.splitext(os.path.basename(path))[0]
    dataset_dir = os.path.join(OUTPUT_DIR, name)
    os.makedirs(dataset_dir, exist_ok=True)

    df = load_dataset(path)

    # 기본 확인
    assert SMILES_COL in df.columns, f"{path}: missing {SMILES_COL}"
    assert LABEL_COL in df.columns, f"{path}: missing {LABEL_COL}"

    df = prepare_three_views(df)
    df.to_csv(os.path.join(dataset_dir, f"{name}_preprocessed_views.csv"), index=False)

    # Canonical-only / Standardized 분석
    canon_df = build_representation_analysis_df(df, "canonical_smiles", "canonical_only")
    std_df = build_representation_analysis_df(df, "standardized_smiles", "standardized")

    canon_df.to_csv(os.path.join(dataset_dir, f"{name}_canonical_analysis.csv"), index=False)
    std_df.to_csv(os.path.join(dataset_dir, f"{name}_standardized_analysis.csv"), index=False)

    # 양성/음성 비교 (기존)
    canon_bin = summarize_binary_features(canon_df)
    canon_num = summarize_numeric_features(canon_df)
    canon_corr = descriptor_label_correlations(canon_df)
    canon_metal = metal_summary(canon_df)

    std_bin = summarize_binary_features(std_df)
    std_num = summarize_numeric_features(std_df)
    std_corr = descriptor_label_correlations(std_df)
    std_metal = metal_summary(std_df)

    canon_bin.to_csv(os.path.join(dataset_dir, f"{name}_canonical_binary_feature_comparison.csv"), index=False)
    canon_num.to_csv(os.path.join(dataset_dir, f"{name}_canonical_numeric_feature_comparison.csv"), index=False)
    canon_corr.to_csv(os.path.join(dataset_dir, f"{name}_canonical_descriptor_label_correlation.csv"), index=False)
    canon_metal.to_csv(os.path.join(dataset_dir, f"{name}_canonical_metal_summary.csv"), index=False)

    std_bin.to_csv(os.path.join(dataset_dir, f"{name}_standardized_binary_feature_comparison.csv"), index=False)
    std_num.to_csv(os.path.join(dataset_dir, f"{name}_standardized_numeric_feature_comparison.csv"), index=False)
    std_corr.to_csv(os.path.join(dataset_dir, f"{name}_standardized_descriptor_label_correlation.csv"), index=False)
    std_metal.to_csv(os.path.join(dataset_dir, f"{name}_standardized_metal_summary.csv"), index=False)

    # ===== NEW: 양성/음성 각각 작용기 ↔ 물리화학적 특성 상관분석 =====
    canon_fg_desc_corr = intra_group_correlation(canon_df)
    std_fg_desc_corr = intra_group_correlation(std_df)

    canon_fg_desc_corr.to_csv(
        os.path.join(dataset_dir, f"{name}_canonical_intragroup_fg_descriptor_corr.csv"), index=False
    )
    std_fg_desc_corr.to_csv(
        os.path.join(dataset_dir, f"{name}_standardized_intragroup_fg_descriptor_corr.csv"), index=False
    )

    # ===== NEW: 양성/음성 각각 물리화학적 특성 간 상관분석 =====
    canon_desc_desc_corr = intra_group_descriptor_correlation(canon_df)
    std_desc_desc_corr = intra_group_descriptor_correlation(std_df)

    canon_desc_desc_corr.to_csv(
        os.path.join(dataset_dir, f"{name}_canonical_intragroup_descriptor_corr.csv"), index=False
    )
    std_desc_desc_corr.to_csv(
        os.path.join(dataset_dir, f"{name}_standardized_intragroup_descriptor_corr.csv"), index=False
    )

    # 전처리 shift 비교
    shift_df = compare_preprocessing_shift(canon_df, std_df, join_cols=("No", "label"))
    shift_df.to_csv(os.path.join(dataset_dir, f"{name}_canonical_vs_standardized_shift.csv"), index=False)

    # 요약
    # Benigni-Bossa 집약 통계 추가
    canon_valid = canon_df[canon_df["valid"] == 1]
    std_valid = std_df[std_df["valid"] == 1]

    summary = pd.DataFrame([{
        "dataset": name,
        "n_total": len(df),
        "positive_ratio": df[LABEL_COL].mean(),
        "canonical_valid": canon_df["valid"].sum(),
        "standardized_valid": std_df["valid"].sum(),
        "canonical_multicomponent_rate": canon_df["is_multicomponent"].mean(),
        "standardized_multicomponent_rate": std_df["is_multicomponent"].mean(),
        "canonical_contains_metal_rate": canon_df["contains_metal"].mean(),
        "standardized_contains_metal_rate": std_df["contains_metal"].mean(),
        "canonical_contains_heavy_metal_rate": canon_df["contains_heavy_metal"].mean(),
        "standardized_contains_heavy_metal_rate": std_df["contains_heavy_metal"].mean(),
        # Benigni-Bossa 요약
        "canonical_bb_any_alert_rate": canon_valid["bb_any_alert"].mean() if len(canon_valid) else np.nan,
        "canonical_bb_genotox_alert_rate": canon_valid["bb_any_genotox_alert"].mean() if len(canon_valid) else np.nan,
        "canonical_bb_nongenotox_alert_rate": canon_valid["bb_any_nongenotox_alert"].mean() if len(canon_valid) else np.nan,
        "standardized_bb_any_alert_rate": std_valid["bb_any_alert"].mean() if len(std_valid) else np.nan,
        "standardized_bb_genotox_alert_rate": std_valid["bb_any_genotox_alert"].mean() if len(std_valid) else np.nan,
        "standardized_bb_nongenotox_alert_rate": std_valid["bb_any_nongenotox_alert"].mean() if len(std_valid) else np.nan,
    }])
    summary.to_csv(os.path.join(dataset_dir, f"{name}_summary.csv"), index=False)

    return {
        "name": name,
        "summary": summary,
        "canon_bin": canon_bin,
        "canon_num": canon_num,
        "std_bin": std_bin,
        "std_num": std_num,
        "shift": shift_df,
    }


def run_all():
    all_summary = []

    for path in INPUT_FILES:
        result = run_analysis_for_file(path)
        all_summary.append(result["summary"])

    all_summary_df = pd.concat(all_summary, ignore_index=True)
    all_summary_df.to_csv(os.path.join(OUTPUT_DIR, "all_dataset_summary.csv"), index=False)
    print(all_summary_df)


if __name__ == "__main__":
    run_all()
```
